# Inhibitory Schur modulation: normalized with-self vs no-self

The signed state matrices are normalized independently to spectral radius 0.85. The comparison covers block perturbations, effective excitatory Schur dynamics, population responses, motif stability, and I-I chain selection.

## Notebook role in the analysis sequence

**Role:** Companion to `07_inhibitory_modulation.ipynb`.

**Inputs:**
- `matrices/mij_matrix.csv`
- `matrices/mij_netlist.csv`

**Outputs:**
- `outputs/inhibitory_schur_modulation/self_comparison/`

**What this notebook adds:**
- Repeats the inhibitory modulation analysis with and without self-connections.
- Uses matched independent spectral-radius normalization to `0.85`.
- Tests whether inhibitory stabilization and motif effects persist after removing diagonal entries.


```python
from pathlib import Path
from IPython.display import display
from inhibitory_schur_modulation_script import (
    load_mij_data,
    run_side_by_side_analysis,
    save_side_by_side_outputs,
)

MATRIX_PATH = Path('matrices/mij_matrix.csv')
NETLIST_PATH = Path('matrices/mij_netlist.csv')
OUTPUT_DIR = Path('outputs/inhibitory_schur_modulation/self_comparison')
```

## Run the matched comparison

```python
data = load_mij_data(MATRIX_PATH, NETLIST_PATH)
comparison = run_side_by_side_analysis(
    data,
    normalization='spectral_radius',
    target=0.85,
    alpha=0.85,
    n_null=50,
)
display(comparison['normalization'])
```

## Core inhibitory results

```python
display(comparison['perturbation'])
display(comparison['schur_summary'])
display(comparison['response_I'])
display(comparison['response_E'])
```

## Motifs and disinhibitory chains

```python
display(comparison['motif_stability'])
display(comparison['neumann']['selected_orders'])
display(comparison['disinhibitors'].groupby('variant').head(10))
```

## Save comparison tables

```python
save_side_by_side_outputs(OUTPUT_DIR, comparison)
print(f'Saved comparison tables to {OUTPUT_DIR.resolve()}')
```
