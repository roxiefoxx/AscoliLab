# Stage 1: E/I Scaffold — Alternating Greedy Tree
## Hippocampal Connectome Analysis | Ascoli Lab, George Mason University

Extends the Stage 0 excitatory scaffold by interleaving inhibitory modulation at every step.
At each excitatory node the notebook identifies the dominant E→I→E sidechain: which inhibitory
interneuron is most strongly recruited, and which excitatory node it most strongly suppresses.

**Path algorithm — E/I/E/I alternating greedy:**
1. Start at an excitatory node E₀
2. At each E node: record (a) the top E→I partner — the inhibitory node receiving the most negative-weight edge from this E node, and (b) that I node's top I→E target — the excitatory node it suppresses most strongly
3. Step the E chain: follow the highest-weight E→E edge to the next unvisited excitatory node
4. Repeat. Stop when no unvisited excitatory neighbors remain (exhaustion rule — identical to Stage 0)

**Design principles:**
- The E spine is *identical* to Stage 0 — no path is modified by the I branches
- I nodes are *not consumed* — the same interneuron can appear alongside multiple E steps
- I→E targets are restricted to excitatory nodes only (avoids I→I suppression artifacts)
- Inhibitory activation normalised as mean(|E→I weights|) × √n to avoid quadratic scaling

**Analyses:**
1. E/I greedy tree — all 28 starting nodes, E/I/IE interleaved
2. Inhibitory hub rankings — most recruited I nodes + activation scores
3. E node suppression load — inhibitory pressure on each excitatory cell
4. E/I region-level flow heatmaps
5. Terminus comparison confirming E chain unchanged

**Matrix convention:** `W[row, col]` = weight FROM row TO col (pre→post).
Inhibitory weights are negative (signed matrix).

**References:** Hippocampome.org v1.0; Rees et al. (eNeuro 2016); Isaacson & Scanziani (2011)

## 2. Load Weight Matrix & Metadata

Load `w_ij_gaa.csv` and `izhikevich_72_v3.csv`. Apply MEC naming patch at load time.
Remove self-loops. Extract all four E/I submatrices.

````python
W_raw = pd.read_csv(WEIGHT_MATRIX, index_col=0)
W_raw.index   = W_raw.index.str.strip()
W_raw.columns = W_raw.columns.str.strip()
W_raw.index   = [MEC_PATCH.get(n, n) for n in W_raw.index]
W_raw.columns = [MEC_PATCH.get(n, n) for n in W_raw.columns]

ei_df = pd.read_csv(NEURON_PARAMS)
ei_df['Neuron Type'] = ei_df['Neuron Type'].str.strip().replace(MEC_PATCH)
ei_df['E/I'] = ei_df['E/I'].str.strip().str.lower()
ei_map = dict(zip(ei_df['Neuron Type'], ei_df['E/I']))

all_nodes  = list(W_raw.index)
excitatory = [n for n in all_nodes if ei_map.get(n, 'i') == 'e']
inhibitory = [n for n in all_nodes if ei_map.get(n, 'i') == 'i']

W = W_raw.copy()
for nd in W.index: W.loc[nd, nd] = 0.0   # remove self-loops

# ── Four E/I submatrices ──────────────────────────────────────────────────────
# W[row, col] = FROM row TO col
# W_EE: E→E (positive weights)
# W_EI: E→I  — W_EI[i_row, e_col] = weight of e→i  (negative, stored in I-row)
# W_IE: I→E  — W_IE[e_row, i_col] = weight of i→e  (negative, stored in E-row)
# W_II: I→I  (negative)

W_EE = W.loc[excitatory, excitatory].clip(lower=0)
W_EE_noself = W_EE.copy()
for nd in W_EE_noself.index: W_EE_noself.loc[nd, nd] = 0.0

W_EI = W.loc[inhibitory, excitatory]   # rows=I, cols=E → W[i,e] = e→i weight
W_IE = W.loc[excitatory, inhibitory]   # rows=E, cols=I → W[e,i] = i→e weight
W_II = W.loc[inhibitory, inhibitory]

G_EE = nx.DiGraph()
G_EE.add_nodes_from(excitatory)
for pre in excitatory:
    for post in excitatory:
        if pre == post: continue
        w = W_EE_noself.loc[pre, post]
        if w > 0: G_EE.add_edge(pre, post, weight=float(w))

print(f'Total nodes   : {len(all_nodes)}')
print(f'Excitatory    : {len(excitatory)}')
print(f'Inhibitory    : {len(inhibitory)}')
print(f'E→E edges     : {G_EE.number_of_edges()}')
print(f'E→I non-zero  : {int((W_EI.values != 0).sum())}')
print(f'I→E non-zero  : {int((W_IE.values != 0).sum())}')
print(f'I→I non-zero  : {int((W_II.values != 0).sum())}')
````

## 3. E/I Path Helpers

Two lookup functions used at every step of the greedy walk:

**`top_EI_target(e_node)`** — finds the inhibitory node that receives the most
strongly negative (most inhibitory) edge from this excitatory node.
`W_EI[i_row, e_col]` stores the E→I weight as a negative value; `idxmin()` selects
the most negative row, i.e. the most strongly driven I node.

**`top_IE_target(i_node)`** — from that I node, finds the excitatory node that
receives the most strongly negative edge back. Restricted to excitatory targets
(avoids I→I suppression artifacts identified in prior Stage 1 debugging).


````python
def top_EI_target(e_node):
    """
    Most strongly activated I node from e_node.
    W_EI[i_row, e_col] = weight of e_node -> i_node (negative).
    idxmin() returns the I node receiving the largest absolute inhibitory drive.
    Returns (i_node, weight) or (None, None) if no E→I edges exist.
    """
    col    = W_EI[e_node]
    col_nz = col[col != 0]
    if col_nz.empty: return None, None
    i_node = col_nz.idxmin()      # most negative = strongest drive
    return i_node, float(col_nz[i_node])


def top_IE_target(i_node):
    """
    Most strongly suppressed E node from i_node.
    W_IE[e_row, i_col] = weight of i_node -> e_node (negative).
    Restricted to excitatory targets only.
    Returns (e_target, weight) or (None, None) if no I→E edges exist.
    """
    col    = W_IE[i_node]
    col_nz = col[col != 0]
    if col_nz.empty: return None, None
    e_target = col_nz.idxmin()    # most negative = strongest suppression
    return e_target, float(col_nz[e_target])


print('Helpers defined: top_EI_target(), top_IE_target()')
print()
# Spot-check on CA3c Pyramidal (dominant out-strength hub)
i_n, ei_w = top_EI_target('CA3c Pyramidal')
ie_n, ie_w = top_IE_target(i_n) if i_n else (None, None)
print(f'CA3c Pyramidal -> top I partner : {i_n}  (w = {ei_w:.3e})')
print(f'{i_n} -> top E suppressed     : {ie_n}  (w = {ie_w:.3e})')
````

## 4. E/I/E/I Alternating Greedy Path

At each E node the path records three things:
1. **E node** — the current excitatory position
2. **I partner** — top E→I target (most strongly recruited interneuron)
3. **IE target** — that interneuron's top I→E output (excitatory cell most suppressed)

Then the E chain steps to the next highest-weight unvisited excitatory neighbor.

**Stopping rule:** identical to Stage 0 — halt when no unvisited excitatory neighbor
remains. I nodes are never consumed and do not affect the stopping condition.


````python
def ei_greedy_path(G_EE, start):
    """
    Build E/I/E/I alternating greedy path from `start`.

    Returns list of step dicts with keys:
      step, e_node, e_step_weight,
      i_node, ei_weight, ie_node, ie_weight,
      stop_reason, would_go_to
    """
    steps, visited, current, step_idx = [], {start}, start, 0

    while True:
        # ── I branch at current E node ────────────────────────────────────────
        i_node, ei_w   = top_EI_target(current)
        ie_node, ie_w  = top_IE_target(i_node) if i_node else (None, None)

        # ── Next E step ───────────────────────────────────────────────────────
        out       = [(v, d['weight']) for _, v, d in G_EE.out_edges(current, data=True)]
        unvisited = [(v, w) for v, w in out if v not in visited]

        if not unvisited:
            would_go = max(out, key=lambda x: x[1])[0] if out else None
            stop     = 'dead_end' if not out else 'exhausted_unvisited_neighbors'
            steps.append({'step':step_idx, 'e_node':current, 'e_step_weight':None,
                          'i_node':i_node, 'ei_weight':ei_w,
                          'ie_node':ie_node, 'ie_weight':ie_w,
                          'stop_reason':stop, 'would_go_to':would_go})
            break

        best_v, best_w = max(unvisited, key=lambda x: x[1])
        steps.append({'step':step_idx, 'e_node':current, 'e_step_weight':best_w,
                      'i_node':i_node, 'ei_weight':ei_w,
                      'ie_node':ie_node, 'ie_weight':ie_w,
                      'stop_reason':None, 'would_go_to':None})
        visited.add(best_v); current = best_v; step_idx += 1

    return steps


# ── Run all 28 starting nodes ─────────────────────────────────────────────────
all_paths   = {s: ei_greedy_path(G_EE, s) for s in excitatory}
path_summary = []

for start, steps in all_paths.items():
    e_chain   = [s['e_node'] for s in steps]
    i_nodes   = [s['i_node'] for s in steps]
    ie_nodes  = [s['ie_node'] for s in steps]
    e_weights = [s['e_step_weight'] for s in steps if s['e_step_weight'] is not None]
    last      = steps[-1]
    path_summary.append({
        'start'           : start,
        'start_region'    : region_of(start),
        'n_e_steps'       : len(steps) - 1,
        'terminus'        : last['e_node'],
        'terminus_region' : region_of(last['e_node']),
        'would_go_to'     : last['would_go_to'],
        'stop_reason'     : last['stop_reason'],
        'total_e_weight'  : sum(e_weights),
        'n_i_branches'    : sum(1 for n in i_nodes if n is not None),
        'n_ie_targets'    : sum(1 for n in ie_nodes if n is not None),
        'e_chain_str'     : ' -> '.join(e_chain),
    })

paths_df = pd.DataFrame(path_summary)
paths_df.to_csv(f'{TABLE_DIR}stage1_paths_summary.csv', index=False)

print(f'Paths traced  : {len(paths_df)}')
print(f'E-step range  : {paths_df["n_e_steps"].min()}–{paths_df["n_e_steps"].max()}')
print(f'I-branch cover: {(paths_df["n_i_branches"]>0).sum()}/{len(paths_df)} paths')
print(f'\nTerminus distribution:')
print(paths_df['terminus'].value_counts().to_string())
print(f'\nWould-go-to distribution:')
print(paths_df['would_go_to'].value_counts().to_string())
````

## 5. E/I/E/I Greedy Tree — All 28 Starting Nodes

### Figure 1 — E/I/E/I Alternating Greedy Tree

Each row is one starting excitatory node (sorted by path length, longest at top).
Each E step occupies three x-slots: **E node** (region colour, large circle) →
**I partner** (red inverted triangle, top E→I) → **IE target** (salmon triangle,
top I→E suppressed E node). Gold star marks the terminus.

**Reading the tree:** Follow a row left-to-right. Large coloured circles are excitatory
relay nodes. At each E node, look directly right: the red ▼ is the interneuron most
strongly activated by that E node, and the salmon ▲ below it is the excitatory cell
that interneuron most strongly suppresses. These three-node units (E→I→E) are the
fundamental inhibitory sidechain motif of the hippocampal excitatory scaffold.


````python
paths_df_s = paths_df.sort_values('n_e_steps', ascending=False).reset_index(drop=True)
max_e_steps = paths_df['n_e_steps'].max()
SLOT_W     = 3
total_cols = SLOT_W * (max_e_steps + 1)

fig_w = max(30, total_cols * 0.9)
fig_h = max(14, len(excitatory) * 0.62)
fig, ax = plt.subplots(figsize=(fig_w, fig_h))
ax.set_facecolor('#12131A'); fig.patch.set_facecolor('#12131A')

for row_idx, row in paths_df_s.iterrows():
    start = row['start']
    steps = all_paths[start]
    y     = -row_idx

    for s in steps:
        col_base = s['step'] * SLOT_W
        e_node, i_node, ie_node = s['e_node'], s['i_node'], s['ie_node']
        is_term = s['stop_reason'] is not None
        xE = col_base

        # E dot
        ec = REGION_COLORS.get(region_of(e_node), '#7F8C8D')
        ax.scatter(xE, y, color=ec, s=(260 if is_term else 140), zorder=4,
                   edgecolors=('gold' if is_term else 'white'),
                   linewidths=(1.8 if is_term else 0.5),
                   marker=('*' if is_term else 'o'))
        ax.text(xE, y + 0.20, e_node.split()[-1][:9], ha='center', va='bottom',
                fontsize=4.8, color='white', fontweight='bold', zorder=5)

        # E→E arrow
        if not is_term and s['e_step_weight']:
            nxt_steps = [st for st in steps if st['step'] == s['step']+1]
            if nxt_steps:
                ax.annotate('', xy=((s['step']+1)*SLOT_W - 0.35, y),
                            xytext=(xE + 0.35, y),
                            arrowprops=dict(arrowstyle='->', color='white', lw=0.8), zorder=3)

        # I dot (E→I)
        if i_node is not None:
            xI = col_base + 1
            ax.scatter(xI, y, color=I_DOT_COLOR, s=70, zorder=4,
                       edgecolors='white', linewidths=0.4, marker='v')
            ax.text(xI, y + 0.18, i_node.split()[-1][:8], ha='center', va='bottom',
                    fontsize=4.0, color=I_DOT_COLOR, zorder=5)
            ax.plot([xE+0.3, xI-0.2], [y, y], color=I_DOT_COLOR,
                    linewidth=0.5, linestyle='--', alpha=0.7, zorder=3)

            # IE dot (I→E)
            if ie_node is not None:
                xIE = col_base + 2
                ax.scatter(xIE, y, color=IE_DOT_COLOR, s=45, zorder=4,
                           edgecolors='white', linewidths=0.3, marker='^')
                ax.text(xIE, y - 0.22, ie_node.split()[-1][:8], ha='center', va='top',
                        fontsize=4.0, color=IE_DOT_COLOR, zorder=5)
                ax.plot([xI+0.2, xIE-0.2], [y, y], color=IE_DOT_COLOR,
                        linewidth=0.5, linestyle=':', alpha=0.6, zorder=3)

    short_s = ' '.join(start.split()[-2:])[:16]
    ax.text(-0.8, y, f"[{region_of(start)}] {short_s}", ha='right', va='center',
            fontsize=6, color=REGION_COLORS.get(region_of(start), 'white'))

e_tick_x = [i * SLOT_W for i in range(max_e_steps + 2)]
ax.set_xticks(e_tick_x)
ax.set_xticklabels([f'E{i}' for i in range(len(e_tick_x))], fontsize=7, color='white')
ax.tick_params(colors='white')
ax.set_xlim(-3.5, total_cols + 0.5)
ax.set_ylim(-len(excitatory) + 0.3, 1.2)
ax.set_yticks([])
for sp in ['top','left','right']: ax.spines[sp].set_visible(False)
ax.spines['bottom'].set_color('#444')
ax.set_xlabel('E step (each slot: E node  ▸  I partner  ▸  I→E target)',
              fontsize=9, color='white')

legend_patches  = [mpatches.Patch(color=REGION_COLORS[r], label=r)
                   for r in ['DG','CA3','CA2','CA1','MEC','LEC','EC']]
legend_patches += [mpatches.Patch(color=I_DOT_COLOR,  label='▼ I node (top E→I)'),
                   mpatches.Patch(color=IE_DOT_COLOR, label='▲ I→E target (suppressed E)'),
                   mpatches.Patch(color='gold',       label='★ E terminus')]
ax.legend(handles=legend_patches, loc='upper right', fontsize=7,
          facecolor='#1E1E2E', labelcolor='white', framealpha=0.9, ncol=2)
ax.set_title('Stage 1: E/I/E/I Alternating Greedy Tree — All 28 Starting Nodes\n'
             '● E node (region colour)  ▼ top E→I interneuron (red)  '
             '▲ top I→E suppressed target (salmon)  ★ terminus',
             fontsize=10, color='white', fontweight='bold', pad=10)
plt.tight_layout()
fig.savefig(f'{TABLE_DIR}fig1_ei_greedy_tree.png', dpi=150, bbox_inches='tight',
            facecolor='#12131A')
plt.show()
````

## 6. Inhibitory Node Activation

**Activation formula (mean × √n):**

For each inhibitory node i:  
`activation(i) = mean(|W_EI[i, :]| for nonzero) × √(n excitatory inputs)`

This is analogous to standard-error scaling. Raw sum would inflate activation
quadratically for high-in-degree I nodes — a DG interneuron receiving from 17 E
nodes would score 17× higher than one receiving from 1, even if the individual
weights are identical. Mean × √n gives a biologically grounded estimate of the
population-average excitatory drive scaled by its reliability (more inputs →
more reliable activation → √n boost).


````python
i_activation = {}
for i_node in inhibitory:
    incoming = W_EI.loc[i_node]
    nz = incoming[incoming != 0]
    if len(nz) == 0:
        i_activation[i_node] = 0.0
    else:
        i_activation[i_node] = float(np.mean(np.abs(nz))) * np.sqrt(len(nz))

i_act_df = pd.DataFrame({
    'i_node'    : inhibitory,
    'region'    : [region_of(n) for n in inhibitory],
    'activation': [i_activation[n] for n in inhibitory],
    'n_e_inputs': [(W_EI.loc[n] != 0).sum() for n in inhibitory],
    'mean_abs_w': [float(np.mean(np.abs(W_EI.loc[n][W_EI.loc[n]!=0])))
                   if (W_EI.loc[n]!=0).any() else 0.0 for n in inhibitory],
}).set_index('i_node').sort_values('activation', ascending=False)
i_act_df.to_csv(f'{TABLE_DIR}stage1_i_activation.csv')

print('Top 10 I nodes by activation (mean×√n):')
print(i_act_df.head(10)[['region','activation','n_e_inputs','mean_abs_w']].to_string())
````

````python
# Count appearances as top E→I partner across all path steps
step_rows = []
for start, steps in all_paths.items():
    for s in steps:
        step_rows.append({'start':start, 'i_node':s['i_node'],
                          'ie_node':s['ie_node']})
steps_df = pd.DataFrame(step_rows)
i_appearances  = steps_df['i_node'].dropna().value_counts()
ie_appearances = steps_df['ie_node'].dropna().value_counts()

i_hub_df = pd.DataFrame({
    'i_node'             : inhibitory,
    'region'             : [region_of(n) for n in inhibitory],
    'top_ei_appearances' : [i_appearances.get(n,0)  for n in inhibitory],
    'top_ie_appearances' : [ie_appearances.get(n,0) for n in inhibitory],
    'ie_out_strength'    : [abs(W_IE[n]).sum() for n in inhibitory],
    'ei_in_strength'     : [abs(W_EI.loc[n]).sum() for n in inhibitory],
    'activation'         : [i_activation[n] for n in inhibitory],
}).set_index('i_node').sort_values('top_ei_appearances', ascending=False)
i_hub_df.to_csv(f'{TABLE_DIR}stage1_i_hub_rankings.csv')

top_n = 15
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Panel A: appearances as top E→I
top_i = i_hub_df.nlargest(top_n,'top_ei_appearances').reset_index()
df_a  = top_i.sort_values('top_ei_appearances')
colors_a = [REGION_COLORS.get(region_of(n),'#7F8C8D') for n in df_a['i_node']]
bars = axes[0].barh(range(len(df_a)), df_a['top_ei_appearances'],
                    color=colors_a, edgecolor='white', linewidth=0.4)
axes[0].set_yticks(range(len(df_a)))
axes[0].set_yticklabels([n[:36] for n in df_a['i_node']], fontsize=9)
axes[0].set_xlabel('Appearances as top E→I partner\n(across all paths × all steps)',fontsize=9)
axes[0].set_title('Most Recruited I Nodes\n(top E→I partner at each path step)',
                   fontsize=10,fontweight='bold')
for bar,v in zip(bars,df_a['top_ei_appearances']):
    axes[0].text(bar.get_width()+0.5,bar.get_y()+bar.get_height()/2,
                 str(int(v)),va='center',fontsize=9)
axes[0].grid(axis='x',alpha=0.25);axes[0].set_axisbelow(True)
for ytick,nd in zip(axes[0].get_yticklabels(),df_a['i_node']):
    ytick.set_color(REGION_COLORS.get(region_of(nd),'#333'))

# Panel B: activation
top_act = i_act_df.nlargest(top_n,'activation').reset_index()
df_b    = top_act.sort_values('activation')
colors_b = [REGION_COLORS.get(region_of(n),'#7F8C8D') for n in df_b['i_node']]
axes[1].barh(range(len(df_b)), df_b['activation'],
             color=colors_b, edgecolor='white', linewidth=0.4)
axes[1].set_yticks(range(len(df_b)))
axes[1].set_yticklabels([n[:36] for n in df_b['i_node']], fontsize=9)
axes[1].set_xlabel('Activation  =  mean(|E→I weights|) × √(n inputs)',fontsize=9)
axes[1].set_title('I Node Activation\n(mean×√n normalisation)',
                   fontsize=10,fontweight='bold')
axes[1].grid(axis='x',alpha=0.25);axes[1].set_axisbelow(True)
for ytick,nd in zip(axes[1].get_yticklabels(),df_b['i_node']):
    ytick.set_color(REGION_COLORS.get(region_of(nd),'#333'))
legend_patches=[mpatches.Patch(color=REGION_COLORS[r],label=r)
                for r in ['DG','CA3','CA2','CA1','MEC','LEC','EC']]
axes[1].legend(handles=legend_patches,fontsize=8,loc='lower right',title='Region')
plt.suptitle('Stage 1: Inhibitory Hub Rankings',fontsize=12,fontweight='bold')
plt.tight_layout()
fig.savefig(f'{TABLE_DIR}fig2_i_hub_rankings.png',dpi=150,bbox_inches='tight')
plt.show()
````

## 7. E-Node Suppression Load

For each excitatory node, compute the total inhibitory pressure it receives:

```
suppression(E) = mean(|W_IE[E,i]| × activation(i) for nonzero i) × √(n_I_inputs)
```

This applies the same mean × √n normalisation to the E-node perspective: how much
inhibitory drive is converging on this excitatory cell, weighted by how strongly
each incoming I node is itself activated. Nodes with high suppression AND high
excitatory out-strength have a high E/I ratio — they sit at a significant
push-pull balance point in the circuit.


````python
suppression = {}
for e_node in excitatory:
    incoming = W_IE.loc[e_node]       # Series indexed by I nodes
    nz = incoming[incoming != 0]
    if nz.empty:
        suppression[e_node] = 0.0
    else:
        weighted = [abs(nz[i]) * i_activation[i] for i in nz.index]
        suppression[e_node] = float(np.mean(weighted)) * np.sqrt(len(nz))

supp_df = pd.DataFrame({
    'e_node'         : excitatory,
    'region'         : [region_of(n) for n in excitatory],
    'suppression'    : [suppression[n] for n in excitatory],
    'n_i_inputs'     : [(W_IE.loc[n] != 0).sum() for n in excitatory],
    'out_strength_EE': [float(W_EE_noself.loc[n].sum()) for n in excitatory],
}).set_index('e_node').sort_values('suppression', ascending=False)
supp_df['ei_ratio'] = supp_df['suppression'] / (supp_df['out_strength_EE'] + 1e-10)
supp_df.to_csv(f'{TABLE_DIR}stage1_e_suppression.csv')

print('Top 10 E nodes by inhibitory suppression received:')
print(supp_df.head(10)[['region','suppression','n_i_inputs',
                         'out_strength_EE','ei_ratio']].to_string())
````

### Figure 3 — E-Node Suppression Load

````python
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

df_s = supp_df.reset_index().sort_values('suppression', ascending=True)
colors_s = [REGION_COLORS.get(region_of(n),'#7F8C8D') for n in df_s['e_node']]
axes[0].barh(range(len(df_s)), np.log10(df_s['suppression']+1),
             color=colors_s, edgecolor='white', linewidth=0.4)
axes[0].set_yticks(range(len(df_s)))
axes[0].set_yticklabels([n[:34] for n in df_s['e_node']], fontsize=9)
axes[0].set_xlabel('log₁₀(suppression score + 1)', fontsize=9)
axes[0].set_title('Inhibitory Suppression per E Node\n'
                   '(weighted by I activation, mean×√n)',fontsize=10,fontweight='bold')
axes[0].grid(axis='x',alpha=0.25); axes[0].set_axisbelow(True)
for ytick,nd in zip(axes[0].get_yticklabels(),df_s['e_node']):
    ytick.set_color(REGION_COLORS.get(region_of(nd),'#333'))

df_r = supp_df.reset_index().sort_values('ei_ratio', ascending=True)
colors_r = [REGION_COLORS.get(region_of(n),'#7F8C8D') for n in df_r['e_node']]
axes[1].barh(range(len(df_r)), np.log10(df_r['ei_ratio']+1),
             color=colors_r, edgecolor='white', linewidth=0.4)
axes[1].set_yticks(range(len(df_r)))
axes[1].set_yticklabels([n[:34] for n in df_r['e_node']], fontsize=9)
axes[1].set_xlabel('log₁₀(suppression / out-strength + 1)', fontsize=9)
axes[1].set_title('E/I Load Ratio per E Node\n'
                   '(suppression ÷ excitatory out-strength)',fontsize=10,fontweight='bold')
axes[1].grid(axis='x',alpha=0.25); axes[1].set_axisbelow(True)
for ytick,nd in zip(axes[1].get_yticklabels(),df_r['e_node']):
    ytick.set_color(REGION_COLORS.get(region_of(nd),'#333'))
legend_patches=[mpatches.Patch(color=REGION_COLORS[r],label=r)
                for r in ['DG','CA3','CA2','CA1','MEC','LEC','EC']]
axes[1].legend(handles=legend_patches,fontsize=8,loc='lower right',title='Region')
plt.suptitle('Stage 1: Inhibitory Suppression on Excitatory Nodes',
             fontsize=12,fontweight='bold')
plt.tight_layout()
fig.savefig(f'{TABLE_DIR}fig3_e_suppression.png',dpi=150,bbox_inches='tight')
plt.show()
````

## 8. E/I Region-Level Weight Flow

### Figure 4 — E/I Flow Heatmaps

Two heatmaps characterise the spatial layout of inhibitory circuitry:
- **E→I flow:** total signed weight delivered FROM each E region TO each I region
- **I→E flow:** total signed weight delivered FROM each I region TO each E region

These are weight-summed (not step-counted), so they reflect the magnitude of
synaptic drive rather than the topology of the greedy path.


````python
region_list = ['EC','MEC','LEC','DG','CA3','CA2','CA1']
ei_flow = pd.DataFrame(0.0, index=region_list, columns=region_list)
ie_flow = pd.DataFrame(0.0, index=region_list, columns=region_list)

for i_node in inhibitory:
    r_i = region_of(i_node)
    if r_i not in region_list: continue
    for e_node, w in W_EI.loc[i_node].items():
        if w != 0:
            r_e = region_of(e_node)
            if r_e in region_list: ei_flow.loc[r_i, r_e] += abs(w)
    for e_node, w in W_IE[i_node].items():
        if w != 0:
            r_e = region_of(e_node)
            if r_e in region_list: ie_flow.loc[r_i, r_e] += abs(w)

ei_flow.to_csv(f'{TABLE_DIR}stage1_ei_region_flow.csv')
ie_flow.to_csv(f'{TABLE_DIR}stage1_ie_region_flow.csv')

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
for ax, mat, title, xlabel, ylabel in [
    (axes[0], ei_flow,
     'E→I Weight Flow\n(total |E→I weight| from E column to I row)',
     'E source region', 'I target region'),
    (axes[1], ie_flow,
     'I→E Weight Flow\n(total |I→E weight| from I row to E column)',
     'E target region', 'I source region'),
]:
    vals = mat.values.astype(float)
    safe = np.where(vals > 0, vals, np.nan)
    im = ax.imshow(safe, cmap='RdPu', aspect='auto',
                   norm=LogNorm(vmin=np.nanmin(safe[~np.isnan(safe)]),
                                vmax=np.nanmax(safe)))
    plt.colorbar(im, ax=ax, label='|weight| (log scale)', fraction=0.04)
    ax.set_xticks(range(len(region_list))); ax.set_xticklabels(region_list,fontsize=10)
    ax.set_yticks(range(len(region_list))); ax.set_yticklabels(region_list,fontsize=10)
    ax.set_xlabel(xlabel, fontsize=9); ax.set_ylabel(ylabel, fontsize=9)
    ax.set_title(title, fontsize=10, fontweight='bold')
    for i in range(len(region_list)):
        for j in range(len(region_list)):
            v = vals[i,j]
            if v > 0:
                c = 'white' if v > np.nanmax(vals)*0.5 else 'black'
                ax.text(j,i,f'{v:.1e}',ha='center',va='center',fontsize=7,color=c)

plt.suptitle('Stage 1: E/I Weight Flow by Region',fontsize=12,fontweight='bold')
plt.tight_layout()
fig.savefig(f'{TABLE_DIR}fig4_ei_region_flow.png',dpi=150,bbox_inches='tight')
plt.show()
````

## 11. Summary

| Component | Finding |
|-----------|---------|
| E-chain structure | Identical to Stage 0 — 28 paths, 6–17 steps, same three attractor basins |
| I-branch coverage | 28/28 paths have at least one E→I partner at every step |
| Most recruited I node | MEC LIII Superficial Trilayered Interneuron (127 appearances across all paths) |
| Highest activation I node | DG Basket (activation 30,832; single strong E→I edge) |
| Highest suppression E node | CA3c Pyramidal — receives inhibitory pressure from 37 I nodes |
| Highest E/I load ratio | DG Mossy — suppression vastly exceeds its excitatory out-strength (ratio 30,449) |
| E/I flow asymmetry | DG and CA3 are the dominant E→I weight donors; CA3 and CA1 are dominant I→E targets |

**Biological interpretation of MEC LIII Superficial Trilayered Interneuron dominance:**  
This node appears as the top E→I partner in 127 of the path steps (across all 28 paths).
It has 20 excitatory input connections, giving it broad coverage across EC and MEC excitatory
populations. Its dominance as a recruited interneuron reflects not raw synaptic strength but
connectivity breadth — it receives input from more E nodes than any other I node in the scaffold.
