# Corrected greedy-tree analysis for the 72-node hippocampal matrix

This notebook uses two source files:

- `w_ij_gaa.xlsx`, sheet `72_wij`: signed weighted 72 × 72 connectivity matrix.
- `izhikevich_72_v3.csv`: authoritative node metadata, including `Neuron Type` and `E/I`.

No E/I identities are inferred from names. The notebook aligns the matrix labels directly to `Neuron Type` and uses the file-provided `E/I` column.

```python
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx

DATA_DIR = Path('/mnt/data')
W_PATH = DATA_DIR / 'w_ij_gaa.xlsx'
META_PATH = DATA_DIR / 'izhikevich_72_v3.csv'
OUTDIR = DATA_DIR / 'greedy_tree_outputs_v2'
FIGDIR = OUTDIR / 'figures'
OUTDIR.mkdir(parents=True, exist_ok=True)
FIGDIR.mkdir(parents=True, exist_ok=True)

W_SHEET = '72_wij'
ROW_IS_SOURCE = True  # If your convention is W[target, source], set this to False.

print('Output directory:', OUTDIR)
```

## 1. Load and align the matrix with authoritative E/I identities

```python
# Load signed weight matrix
W_df_raw = pd.read_excel(W_PATH, sheet_name=W_SHEET, index_col=0)
W_df_raw = W_df_raw.apply(pd.to_numeric, errors='coerce').fillna(0.0)

# Ensure same row/column ordering if labels match
if set(W_df_raw.index) != set(W_df_raw.columns):
    raise ValueError('The row and column labels in the weight matrix do not match.')
W_df = W_df_raw.loc[list(W_df_raw.index), list(W_df_raw.index)].copy()
np.fill_diagonal(W_df.values, 0.0)

# Load metadata
meta = pd.read_csv(META_PATH)
required = {'Neuron Type', 'E/I'}
missing_cols = required - set(meta.columns)
if missing_cols:
    raise ValueError(f'Metadata file is missing required columns: {missing_cols}')

meta = meta.copy()
meta['Neuron Type'] = meta['Neuron Type'].astype(str).str.strip()
meta['E/I_raw'] = meta['E/I'].astype(str).str.strip()
meta['E/I'] = meta['E/I_raw'].str.upper().str[0]

valid_ei = {'E', 'I'}
bad_ei = sorted(set(meta['E/I']) - valid_ei)
if bad_ei:
    raise ValueError(f'Unexpected E/I labels after normalization: {bad_ei}')

# Validate label matching
matrix_nodes = list(W_df.index)
meta_nodes = set(meta['Neuron Type'])
missing_from_meta = sorted(set(matrix_nodes) - meta_nodes)
extra_in_meta = sorted(meta_nodes - set(matrix_nodes))

print('Matrix shape:', W_df.shape)
print('Metadata shape:', meta.shape)
print('Missing matrix labels from metadata:', len(missing_from_meta))
print('Extra metadata labels not in matrix:', len(extra_in_meta))
if missing_from_meta:
    print('Missing from metadata:', missing_from_meta)
if extra_in_meta:
    print('Extra in metadata:', extra_in_meta[:10])

# Known one-label naming mismatch between the matrix and metadata file.
# This does not infer E/I from keywords; it maps the matrix label to the corresponding metadata row.
LABEL_ALIASES = {
    'MEC LV Multipolar Pyramidal': 'MEC LIII Multipolar Principal',
}

meta_lookup = meta.set_index('Neuron Type')['E/I'].to_dict()
unresolved = [n for n in matrix_nodes if n not in meta_lookup and n not in LABEL_ALIASES]
if unresolved:
    raise ValueError(f'Cannot proceed: unresolved matrix labels absent from metadata: {unresolved}')

node_df = pd.DataFrame({'node': matrix_nodes})
node_df['metadata_label'] = node_df['node'].map(lambda x: LABEL_ALIASES.get(x, x))
node_df['E/I'] = node_df['metadata_label'].map(meta_lookup)
if node_df['E/I'].isna().any():
    raise ValueError('Some E/I identities could not be mapped after applying LABEL_ALIASES.')
if LABEL_ALIASES:
    print('Applied label aliases:', LABEL_ALIASES)
node_df['region'] = node_df['node'].str.extract(r'^(DG|CA1|CA2|CA3|EC|MEC|LEC)', expand=False).fillna('Other')
node_df.to_csv(OUTDIR / 'node_metadata_aligned.csv', index=False)

print('\nE/I counts from izhikevich_72_v3.csv:')
print(node_df['E/I'].value_counts())
node_df.head()
```

## 2. Basic signed-weight network summary

```python
W = W_df.to_numpy(float)
labels = list(W_df.index)
ei = dict(zip(node_df['node'], node_df['E/I']))
regions = dict(zip(node_df['node'], node_df['region']))

source_labels = labels if ROW_IS_SOURCE else labels
target_labels = labels if ROW_IS_SOURCE else labels

edge_rows = []
for i, src in enumerate(labels):
    for j, tgt in enumerate(labels):
        if i == j:
            continue
        w = W[i, j] if ROW_IS_SOURCE else W[j, i]
        if w != 0:
            edge_rows.append({
                'source': src, 'target': tgt, 'weight': w, 'abs_weight': abs(w),
                'source_EI': ei[src], 'target_EI': ei[tgt],
                'source_region': regions[src], 'target_region': regions[tgt],
                'edge_sign': 'positive' if w > 0 else 'negative'
            })
edges_df = pd.DataFrame(edge_rows)
edges_df.to_csv(OUTDIR / 'all_nonzero_edges.csv', index=False)

summary = pd.Series({
    'n_nodes': len(labels),
    'n_edges_nonzero': len(edges_df),
    'n_positive_edges': int((edges_df['weight'] > 0).sum()),
    'n_negative_edges': int((edges_df['weight'] < 0).sum()),
    'density_nonzero_no_self': len(edges_df) / (len(labels) * (len(labels)-1)),
    'min_weight': edges_df['weight'].min(),
    'max_weight': edges_df['weight'].max(),
    'median_abs_weight': edges_df['abs_weight'].median(),
    'mean_abs_weight': edges_df['abs_weight'].mean(),
})
summary.to_csv(OUTDIR / 'network_summary.csv')
print(summary)

block_summary = edges_df.groupby(['source_EI','target_EI','edge_sign']).agg(
    n_edges=('weight','size'),
    mean_weight=('weight','mean'),
    median_abs_weight=('abs_weight','median'),
    max_abs_weight=('abs_weight','max')
).reset_index()
block_summary.to_csv(OUTDIR / 'edge_summary_by_EI_block.csv', index=False)
block_summary
```

```python
fig, ax = plt.subplots(figsize=(7,4))
ax.hist(edges_df.loc[edges_df['weight'] > 0, 'weight'], bins=40, alpha=0.7, label='positive')
ax.hist(edges_df.loc[edges_df['weight'] < 0, 'weight'], bins=40, alpha=0.7, label='negative')
ax.set_title('Signed edge-weight distribution')
ax.set_xlabel('Weight')
ax.set_ylabel('Number of directed edges')
ax.legend()
fig.tight_layout()
fig.savefig(FIGDIR / 'signed_weight_distribution.png', dpi=200)
plt.show()
```

## 3. Greedy helper functions

The core routine below builds one greedy path from a start node. At each step, the current node selects the strongest allowable outgoing edge to an unvisited target. For positive-only paths, "strongest" means largest positive weight. For inhibitory-to-excitatory steps in the alternating rule, "strongest inhibition" means most negative weight.

```python
def outgoing_edges(source, allowed_targets=None, sign='positive'):
    sub = edges_df[edges_df['source'].eq(source)].copy()
    if allowed_targets is not None:
        sub = sub[sub['target'].isin(set(allowed_targets))]
    if sign == 'positive':
        sub = sub[sub['weight'] > 0].sort_values('weight', ascending=False)
    elif sign == 'negative':
        sub = sub[sub['weight'] < 0].assign(strength=lambda d: d['weight'].abs()).sort_values('strength', ascending=False)
    elif sign == 'abs':
        sub = sub.assign(strength=lambda d: d['weight'].abs()).sort_values('strength', ascending=False)
    else:
        raise ValueError("sign must be 'positive', 'negative', or 'abs'")
    return sub

def greedy_path(start, allowed_nodes=None, sign='positive', max_steps=None):
    if allowed_nodes is None:
        allowed_nodes = set(labels)
    else:
        allowed_nodes = set(allowed_nodes)
    if max_steps is None:
        max_steps = len(allowed_nodes) - 1

    path = [start]
    visited = {start}
    chosen_edges = []
    current = start
    termination_reason = 'max_steps_reached'

    for _ in range(max_steps):
        candidates = outgoing_edges(current, allowed_targets=allowed_nodes - visited, sign=sign)
        if candidates.empty:
            termination_reason = 'no_allowable_unvisited_outgoing_edge'
            break
        row = candidates.iloc[0].to_dict()
        chosen_edges.append(row)
        current = row['target']
        path.append(current)
        visited.add(current)

    return {
        'start': start,
        'terminal': path[-1],
        'path': path,
        'path_string': ' → '.join(path),
        'n_nodes': len(path),
        'n_edges': len(chosen_edges),
        'total_weight': float(sum(e['weight'] for e in chosen_edges)),
        'total_abs_weight': float(sum(abs(e['weight']) for e in chosen_edges)),
        'termination_reason': termination_reason,
        'edges': chosen_edges
    }

def paths_to_tables(paths, prefix):
    path_rows = []
    edge_rows = []
    for p in paths:
        path_rows.append({k:v for k,v in p.items() if k != 'edges'})
        for step, e in enumerate(p['edges'], start=1):
            edge_rows.append({
                'start': p['start'], 'step': step,
                'source': e['source'], 'target': e['target'], 'weight': e['weight'],
                'source_EI': e['source_EI'], 'target_EI': e['target_EI'],
                'source_region': e['source_region'], 'target_region': e['target_region']
            })
    path_df = pd.DataFrame(path_rows)
    edge_df = pd.DataFrame(edge_rows)
    path_df.to_csv(OUTDIR / f'{prefix}_paths.csv', index=False)
    edge_df.to_csv(OUTDIR / f'{prefix}_chosen_edges.csv', index=False)
    return path_df, edge_df
```

## 4. Excitatory-only positive greedy backbone

This is the clean version of the original excitatory-only backbone: starts are excitatory nodes only, targets are excitatory nodes only, and edges must have positive weight.

```python
E_nodes = node_df.loc[node_df['E/I'].eq('E'), 'node'].tolist()
I_nodes = node_df.loc[node_df['E/I'].eq('I'), 'node'].tolist()

E_only_paths = [greedy_path(start, allowed_nodes=E_nodes, sign='positive') for start in E_nodes]
E_path_df, E_edge_df = paths_to_tables(E_only_paths, 'E_only_positive_greedy')

print('Number of excitatory starts:', len(E_nodes))
print('Mean path length, edges:', E_path_df['n_edges'].mean())
print('Max path length, edges:', E_path_df['n_edges'].max())
print('Most common terminals:')
print(E_path_df['terminal'].value_counts().head(10))
E_path_df.sort_values(['n_edges','total_weight'], ascending=False).head(10)
```

```python
fig, ax = plt.subplots(figsize=(7,4))
ax.hist(E_path_df['n_edges'], bins=range(int(E_path_df['n_edges'].max())+2), align='left')
ax.set_title('Excitatory-only greedy path lengths')
ax.set_xlabel('Path length (edges)')
ax.set_ylabel('Number of E starts')
fig.tight_layout()
fig.savefig(FIGDIR / 'E_only_path_length_histogram.png', dpi=200)
plt.show()

fig, ax = plt.subplots(figsize=(8,5))
term_counts = E_path_df['terminal'].value_counts().head(15).sort_values()
ax.barh(term_counts.index, term_counts.values)
ax.set_title('Most frequent E-only greedy terminal nodes')
ax.set_xlabel('Number of starts ending at node')
fig.tight_layout()
fig.savefig(FIGDIR / 'E_only_terminal_counts.png', dpi=200)
plt.show()
```

```python
# Build transition graph from all chosen E-only greedy edges
G_E = nx.DiGraph()
for _, r in E_edge_df.iterrows():
    if G_E.has_edge(r['source'], r['target']):
        G_E[r['source']][r['target']]['count'] += 1
        G_E[r['source']][r['target']]['weight_sum'] += r['weight']
    else:
        G_E.add_edge(r['source'], r['target'], count=1, weight_sum=r['weight'])

node_visit_counts = pd.Series([n for p in E_only_paths for n in p['path']]).value_counts().rename_axis('node').reset_index(name='visits')
node_visit_counts = node_visit_counts.merge(node_df, on='node', how='left')
node_visit_counts.to_csv(OUTDIR / 'E_only_node_visit_counts.csv', index=False)

print('E-only transition graph nodes:', G_E.number_of_nodes())
print('E-only transition graph edges:', G_E.number_of_edges())
node_visit_counts.head(15)
```

```python
fig, ax = plt.subplots(figsize=(9,5))
top_visits = node_visit_counts.head(20).sort_values('visits')
ax.barh(top_visits['node'], top_visits['visits'])
ax.set_title('Most visited nodes in E-only greedy paths')
ax.set_xlabel('Visit count across all starts')
fig.tight_layout()
fig.savefig(FIGDIR / 'E_only_top_visited_nodes.png', dpi=200)
plt.show()
```

## 5. Signed E/I alternating greedy paths

This optional analysis preserves the biological constraint discussed during the project: inhibitory nodes are entered from excitatory activity, and inhibitory steps are interpreted as suppression of a downstream excitatory target.

Rules used here:

1. Start from each excitatory node.
2. If current node is E, choose the strongest positive outgoing edge to any unvisited node.
3. If current node is I, choose the strongest negative outgoing edge to an unvisited E node.
4. Stop when no allowable unvisited edge remains.

```python
def greedy_alternating_EI_path(start, max_steps=None):
    if ei[start] != 'E':
        raise ValueError('Alternating E/I paths must start from an excitatory node.')
    if max_steps is None:
        max_steps = len(labels) - 1

    path = [start]
    visited = {start}
    chosen_edges = []
    current = start
    termination_reason = 'max_steps_reached'

    for _ in range(max_steps):
        if ei[current] == 'E':
            candidates = outgoing_edges(current, allowed_targets=set(labels)-visited, sign='positive')
        else:
            candidates = outgoing_edges(current, allowed_targets=set(E_nodes)-visited, sign='negative')
        if candidates.empty:
            termination_reason = f'no_allowable_unvisited_outgoing_edge_from_{ei[current]}'
            break
        row = candidates.iloc[0].to_dict()
        chosen_edges.append(row)
        current = row['target']
        path.append(current)
        visited.add(current)

    return {
        'start': start,
        'terminal': path[-1],
        'path': path,
        'path_string': ' → '.join(path),
        'n_nodes': len(path),
        'n_edges': len(chosen_edges),
        'total_weight': float(sum(e['weight'] for e in chosen_edges)),
        'total_abs_weight': float(sum(abs(e['weight']) for e in chosen_edges)),
        'n_E_nodes': sum(ei[n]=='E' for n in path),
        'n_I_nodes': sum(ei[n]=='I' for n in path),
        'termination_reason': termination_reason,
        'edges': chosen_edges
    }

alt_paths = [greedy_alternating_EI_path(start) for start in E_nodes]
ALT_path_df, ALT_edge_df = paths_to_tables(alt_paths, 'EI_alternating_signed_greedy')

print('Mean path length, edges:', ALT_path_df['n_edges'].mean())
print('Max path length, edges:', ALT_path_df['n_edges'].max())
print('Mean inhibitory nodes visited:', ALT_path_df['n_I_nodes'].mean())
ALT_path_df.sort_values(['n_edges','total_abs_weight'], ascending=False).head(10)
```

```python
fig, ax = plt.subplots(figsize=(7,4))
ax.hist(ALT_path_df['n_edges'], bins=range(int(ALT_path_df['n_edges'].max())+2), align='left')
ax.set_title('Signed E/I alternating greedy path lengths')
ax.set_xlabel('Path length (edges)')
ax.set_ylabel('Number of E starts')
fig.tight_layout()
fig.savefig(FIGDIR / 'EI_alternating_path_length_histogram.png', dpi=200)
plt.show()

fig, ax = plt.subplots(figsize=(8,5))
term_counts = ALT_path_df['terminal'].value_counts().head(15).sort_values()
ax.barh(term_counts.index, term_counts.values)
ax.set_title('Most frequent E/I alternating greedy terminal nodes')
ax.set_xlabel('Number of starts ending at node')
fig.tight_layout()
fig.savefig(FIGDIR / 'EI_alternating_terminal_counts.png', dpi=200)
plt.show()
```

## 6. Region-level flow summaries

```python
def region_flow_table(edge_df, prefix):
    if edge_df.empty:
        return pd.DataFrame()
    tbl = edge_df.groupby(['source_region','target_region']).agg(
        n_steps=('weight','size'),
        mean_weight=('weight','mean'),
        total_abs_weight=('weight', lambda x: np.abs(x).sum())
    ).reset_index().sort_values(['n_steps','total_abs_weight'], ascending=False)
    tbl.to_csv(OUTDIR / f'{prefix}_region_flows.csv', index=False)
    return tbl

E_region_flow = region_flow_table(E_edge_df, 'E_only_positive_greedy')
ALT_region_flow = region_flow_table(ALT_edge_df, 'EI_alternating_signed_greedy')

print('Top E-only region flows:')
display(E_region_flow.head(10))
print('Top E/I alternating region flows:')
display(ALT_region_flow.head(10))
```

```python
def plot_region_heatmap(tbl, title, outfile):
    if tbl.empty:
        return
    regions_sorted = sorted(set(tbl['source_region']) | set(tbl['target_region']))
    mat = tbl.pivot(index='source_region', columns='target_region', values='n_steps').reindex(index=regions_sorted, columns=regions_sorted).fillna(0)
    fig, ax = plt.subplots(figsize=(6,5))
    im = ax.imshow(mat.values)
    ax.set_xticks(range(len(regions_sorted)))
    ax.set_yticks(range(len(regions_sorted)))
    ax.set_xticklabels(regions_sorted, rotation=45, ha='right')
    ax.set_yticklabels(regions_sorted)
    ax.set_title(title)
    ax.set_xlabel('Target region')
    ax.set_ylabel('Source region')
    fig.colorbar(im, ax=ax, label='Number of greedy steps')
    fig.tight_layout()
    fig.savefig(FIGDIR / outfile, dpi=200)
    plt.show()

plot_region_heatmap(E_region_flow, 'E-only greedy region flow counts', 'E_only_region_flow_heatmap.png')
plot_region_heatmap(ALT_region_flow, 'E/I alternating greedy region flow counts', 'EI_alternating_region_flow_heatmap.png')
```

## 8. Plain-language interpretation template

Use these outputs as a compact interpretation layer:

- `E_only_positive_greedy_paths.csv`: the excitatory-only scaffold. Each row is one excitatory start node and the path produced by repeatedly taking the strongest positive E→E edge that does not revisit a node.
- `E_only_node_visit_counts.csv`: identifies recurrent hubs in the excitatory scaffold. High visit count means many greedy routes pass through that node.
- `E_only_positive_greedy_region_flows.csv`: summarizes whether the greedy scaffold recovers recognizable region-level routing patterns.
- `EI_alternating_signed_greedy_paths.csv`: a signed extension that allows inhibitory nodes into the walk but forces inhibitory nodes to suppress excitatory targets through the strongest negative I→E edge.

The main methodological correction is that E/I identity is taken from `izhikevich_72_v3.csv`, not inferred from cell-type keywords.
