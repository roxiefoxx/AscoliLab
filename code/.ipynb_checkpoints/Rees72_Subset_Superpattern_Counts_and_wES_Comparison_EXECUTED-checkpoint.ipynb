# Rees 72-Node Subset: Superpattern Count Audit + Rees ES vs wES Comparison

This notebook subsets the Rees-style trimer enumeration table to the neuron labels available in the 72-node weighted `W_ij` matrix, audits whether superpattern counts are preserved, and compares recomputed Rees-style ES against weighted `W_ij` wES.

Inputs expected:
- `w_ij_gaa.csv`
- `stage2a_triad_enumeration_all.csv`


```python
# ============================================================
# 0. Setup: imports, paths, optional Colab support
# ============================================================

from pathlib import Path
import os, sys, warnings
warnings.filterwarnings("ignore")

IN_COLAB = "google.colab" in sys.modules
if IN_COLAB:
    from google.colab import drive
    drive.mount('/content/drive')
    PROJECT_DIR = Path('/content/drive/MyDrive/Ascoli_LabRotation')
else:
    PROJECT_DIR = Path('/mnt/data') if Path('/mnt/data').exists() else Path.cwd()

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    import plotly.express as px
except Exception:
    px = None

OUTPUT_DIR = PROJECT_DIR / 'outputs_rees72_subset_wes_comparison'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

WEIGHT_PATH_CANDIDATES = [
    PROJECT_DIR / 'w_ij_gaa.csv',
    PROJECT_DIR / 'matrices' / 'w_ij_gaa.csv',
    Path('/mnt/data/w_ij_gaa.csv'),
    Path('./w_ij_gaa.csv'),
    Path('../matrices/w_ij_gaa.csv'),
]
ENUM_PATH_CANDIDATES = [
    PROJECT_DIR / 'stage2a_triad_enumeration_all.csv',
    PROJECT_DIR / 'outputs' / 'stage2a_triad_enumeration_all.csv',
    Path('/mnt/data/stage2a_triad_enumeration_all.csv'),
    Path('./stage2a_triad_enumeration_all.csv'),
    Path('../outputs/stage2a_triad_enumeration_all.csv'),
]

WEIGHT_PATH = next((p for p in WEIGHT_PATH_CANDIDATES if p.exists()), None)
ENUM_PATH = next((p for p in ENUM_PATH_CANDIDATES if p.exists()), None)

if WEIGHT_PATH is None:
    raise FileNotFoundError('Could not find w_ij_gaa.csv. Set WEIGHT_PATH manually.')
if ENUM_PATH is None:
    raise FileNotFoundError('Could not find stage2a_triad_enumeration_all.csv. Set ENUM_PATH manually.')

print('PROJECT_DIR :', PROJECT_DIR)
print('OUTPUT_DIR  :', OUTPUT_DIR)
print('WEIGHT_PATH :', WEIGHT_PATH)
print('ENUM_PATH   :', ENUM_PATH)
```

## 1. Load weighted matrix and Rees-style trimer enumeration

The weighted matrix supplies the real signed `W_ij` values. The enumeration table supplies the Rees-style trimer identities, superpattern labels, E/I labels, and binary topology.

```python
# ============================================================
# 1. Load inputs
# ============================================================

A_df = pd.read_csv(WEIGHT_PATH, index_col=0)
A_df = A_df.apply(pd.to_numeric, errors='coerce').fillna(0.0)
A_df.index = A_df.index.astype(str).str.strip()
A_df.columns = A_df.columns.astype(str).str.strip()

if set(A_df.index) == set(A_df.columns):
    A_df = A_df.loc[A_df.index, A_df.index]

if A_df.shape[0] != A_df.shape[1]:
    raise ValueError(f'Weighted matrix is not square: {A_df.shape}')
if list(A_df.index) != list(A_df.columns):
    raise ValueError('Weighted matrix row/column labels do not match.')

W = A_df.to_numpy(dtype=float)
labels = list(A_df.index)

df_enum_raw = pd.read_csv(ENUM_PATH)

print('Weighted matrix shape:', A_df.shape)
print('Enumeration table shape:', df_enum_raw.shape)
print('Enumeration columns:', list(df_enum_raw.columns))
```

## 2. Standardize trimer columns and subset to 72-node `W_ij` namespace

This section maps the enumeration table onto the 72-node weighted matrix labels. If all enumerated trimers already use the same 72-node namespace, the subset count will equal the original enumeration count.

```python
# ============================================================
# 2. Standardize node columns and subset to W72
# ============================================================

df_rees_enum = df_enum_raw.copy()

if all(c in df_rees_enum.columns for c in ['node1', 'node2', 'node3']):
    df_rees_enum = df_rees_enum.rename(columns={'node1': 'node_1', 'node2': 'node_2', 'node3': 'node_3'})
elif all(c in df_rees_enum.columns for c in ['i', 'j', 'k']):
    df_rees_enum = df_rees_enum.rename(columns={'i': 'node_1', 'j': 'node_2', 'k': 'node_3'})
elif all(c in df_rees_enum.columns for c in ['A', 'B', 'C']):
    df_rees_enum = df_rees_enum.rename(columns={'A': 'node_1', 'B': 'node_2', 'C': 'node_3'})

node_cols = ['node_1', 'node_2', 'node_3']
missing = [c for c in node_cols if c not in df_rees_enum.columns]
if missing:
    raise ValueError(f'Missing node columns after standardization: {missing}')

for c in node_cols:
    df_rees_enum[c] = df_rees_enum[c].astype(str).str.strip()
    df_rees_enum[f'{c}_original'] = df_rees_enum[c]

# Known label repairs from prior audit. Extend if needed.
LABEL_REMAP = {
    'MEC LIII Multipolar Principal': 'LEC LIII Multipolar Principal',
    'CA1 Oriens/Alveus': 'CA1 Oriens Alveus',
}
for c in node_cols:
    df_rees_enum[c] = df_rees_enum[c].replace(LABEL_REMAP)

matrix_nodes = set(A_df.index)
for c in node_cols:
    df_rees_enum[f'{c}_in_W72'] = df_rees_enum[c].isin(matrix_nodes)

mask_all_in_W72 = df_rees_enum[[f'{c}_in_W72' for c in node_cols]].all(axis=1)
df_rees_72_subset = df_rees_enum[mask_all_in_W72].copy().reset_index(drop=True)
df_rees_outside_72 = df_rees_enum[~mask_all_in_W72].copy().reset_index(drop=True)

print('Rees/enumeration trimers total:', len(df_rees_enum))
print('Trimers fully mappable to W72:', len(df_rees_72_subset))
print('Trimers excluded due to missing W72 node:', len(df_rees_outside_72))

if len(df_rees_outside_72) > 0:
    outside_nodes = sorted(set(pd.concat([df_rees_outside_72[c] for c in node_cols])) - matrix_nodes)
    print('Unique labels outside W72 namespace:', len(outside_nodes))
    display(pd.DataFrame({'outside_label': outside_nodes}).head(50))

df_rees_72_subset.to_csv(OUTPUT_DIR / 'section2_rees_trimers_subset_to_W72.csv', index=False)
df_rees_outside_72.to_csv(OUTPUT_DIR / 'section2_rees_trimers_outside_W72.csv', index=False)
```

## 3. Define connected Rees A–M subset

Rees connected superpatterns A–M exclude disconnected cases such as `-A`, `-B`, and `-C`. This cell uses the enumeration's `is_connected` column when available. If that column is absent, it recomputes the connected criterion from directed edge presence.

```python
# ============================================================
# 3. Connected Rees A-M subset
# ============================================================

if 'is_connected' in df_rees_72_subset.columns:
    connected_mask = df_rees_72_subset['is_connected'].astype(bool)
elif 'superpattern' in df_rees_72_subset.columns:
    connected_mask = ~df_rees_72_subset['superpattern'].astype(str).str.startswith('-')
else:
    W_bool = W != 0
    np.fill_diagonal(W_bool, False)
    node_to_idx = {n: i for i, n in enumerate(labels)}
    vals = []
    for _, row in df_rees_72_subset.iterrows():
        idx = [node_to_idx[row[c]] for c in node_cols]
        sub = W_bool[np.ix_(idx, idx)]
        n_edges = int(sub.sum())
        touches = sub.any(axis=0) | sub.any(axis=1)
        vals.append(n_edges >= 2 and touches.all())
    connected_mask = pd.Series(vals, index=df_rees_72_subset.index)

df_rees_72_connected_AM = df_rees_72_subset[connected_mask].copy().reset_index(drop=True)
df_rees_72_disconnected = df_rees_72_subset[~connected_mask].copy().reset_index(drop=True)

print('W72 subset trimers:', len(df_rees_72_subset))
print('Connected Rees A-M candidate trimers:', len(df_rees_72_connected_AM))
print('Disconnected/non-A-M trimers:', len(df_rees_72_disconnected))

display(df_rees_72_connected_AM.head())

df_rees_72_connected_AM.to_csv(OUTPUT_DIR / 'section3_rees_connected_A_to_M_subset_W72.csv', index=False)
df_rees_72_disconnected.to_csv(OUTPUT_DIR / 'section3_rees_disconnected_subset_W72.csv', index=False)
```

## 4. Superpattern count audit

This checks whether the superpattern counts from the full enumeration are reproduced after subsetting to the 72-node matrix. If the enumeration is already 72-node specific, the first two count columns should match exactly.

```python
# ============================================================
# 4. Superpattern count audit
# ============================================================

if 'superpattern' not in df_rees_enum.columns:
    raise ValueError('No superpattern column available for count audit.')

full_counts = df_rees_enum['superpattern'].astype(str).value_counts(dropna=False).rename_axis('superpattern').reset_index(name='rees_enum_count')
subset_counts = df_rees_72_subset['superpattern'].astype(str).value_counts(dropna=False).rename_axis('superpattern').reset_index(name='W72_subset_count')
connected_counts = df_rees_72_connected_AM['superpattern'].astype(str).value_counts(dropna=False).rename_axis('superpattern').reset_index(name='W72_connected_AM_count')

superpattern_count_audit = full_counts.merge(subset_counts, on='superpattern', how='outer').merge(connected_counts, on='superpattern', how='outer').fillna(0)
for c in ['rees_enum_count', 'W72_subset_count', 'W72_connected_AM_count']:
    superpattern_count_audit[c] = superpattern_count_audit[c].astype(int)

superpattern_count_audit['subset_minus_rees_enum'] = superpattern_count_audit['W72_subset_count'] - superpattern_count_audit['rees_enum_count']
superpattern_count_audit['connected_AM_minus_rees_enum'] = superpattern_count_audit['W72_connected_AM_count'] - superpattern_count_audit['rees_enum_count']
superpattern_count_audit['subset_matches_rees_enum'] = superpattern_count_audit['subset_minus_rees_enum'] == 0
superpattern_count_audit = superpattern_count_audit.sort_values('superpattern').reset_index(drop=True)

display(superpattern_count_audit)

print('Total Rees/enumeration trimers:', int(superpattern_count_audit['rees_enum_count'].sum()))
print('Total W72 subset trimers:', int(superpattern_count_audit['W72_subset_count'].sum()))
print('Total W72 connected A-M trimers:', int(superpattern_count_audit['W72_connected_AM_count'].sum()))
print('Subset counts match original enumeration counts:', bool(superpattern_count_audit['subset_matches_rees_enum'].all()))

if 'n_edges' in df_rees_enum.columns:
    edge_count_audit = (
        df_rees_enum['n_edges'].value_counts().rename_axis('n_edges').reset_index(name='rees_enum_count')
        .merge(df_rees_72_subset['n_edges'].value_counts().rename_axis('n_edges').reset_index(name='W72_subset_count'), on='n_edges', how='outer')
        .merge(df_rees_72_connected_AM['n_edges'].value_counts().rename_axis('n_edges').reset_index(name='W72_connected_AM_count'), on='n_edges', how='outer')
        .fillna(0).sort_values('n_edges').reset_index(drop=True)
    )
    for c in ['rees_enum_count', 'W72_subset_count', 'W72_connected_AM_count']:
        edge_count_audit[c] = edge_count_audit[c].astype(int)
    display(edge_count_audit)
    edge_count_audit.to_csv(OUTPUT_DIR / 'section4_edge_count_audit_rees_vs_W72.csv', index=False)
else:
    edge_count_audit = pd.DataFrame()

superpattern_count_audit.to_csv(OUTPUT_DIR / 'section4_superpattern_count_audit_rees_vs_W72.csv', index=False)
```

## 5. Recompute Rees ES and weighted `W_ij` wES for the 72-node connected A–M subset

Rees ES is recomputed using categorical E/I labels and binary edge presence. wES uses the same trimer list but replaces binary modulation with real signed `W_ij` magnitudes.

```python
# ============================================================
# 5. Rees ES vs weighted W_ij wES
# ============================================================

ALPHA = 0.10
BETA = 0.10

def normalize_ei_value(x):
    x = str(x).strip().lower()
    if x in {'e', 'exc', 'excitatory', '+', '+1', '1', 'glutamatergic'}:
        return 1
    if x in {'i', 'inh', 'inhibitory', '-', '-1', 'gabaergic'}:
        return -1
    return np.nan

def infer_node_sign_from_outgoing(node):
    return 1 if A_df.loc[node].sum() >= 0 else -1

def get_node_signs(row):
    ei_cols = ['ei_1', 'ei_2', 'ei_3']
    if all(c in row.index for c in ei_cols):
        signs = [normalize_ei_value(row[c]) for c in ei_cols]
        if not any(pd.isna(s) for s in signs):
            return [int(s) for s in signs]
    return [infer_node_sign_from_outgoing(row[c]) for c in node_cols]

def rees_es_from_submatrix(subW, node_signs):
    edge_present = subW != 0
    np.fill_diagonal(edge_present, False)
    node_scores = []
    for target_idx in range(3):
        score = float(node_signs[target_idx])
        for source_idx in range(3):
            if source_idx == target_idx:
                continue
            if edge_present[source_idx, target_idx]:
                score *= 1.1 if node_signs[source_idx] == 1 else 0.9
        node_scores.append(score)
    return float(np.sum(node_scores)), node_scores

def weighted_wES_from_submatrix(subW, node_signs, alpha=ALPHA, beta=BETA):
    node_scores = []
    for target_idx in range(3):
        incoming = subW[:, target_idx].copy()
        incoming[target_idx] = 0.0
        exc_drive = incoming[incoming > 0].sum()
        inh_drive = np.abs(incoming[incoming < 0]).sum()
        s_i = node_signs[target_idx]
        score = s_i * (1 + alpha * np.log1p(exc_drive) - beta * np.log1p(inh_drive))
        node_scores.append(float(score))
    return float(np.sum(node_scores)), node_scores

def spectral_metrics(subW):
    eigvals = np.linalg.eigvals(subW)
    abs_eigs = np.abs(eigvals)
    dom_idx = int(np.argmax(abs_eigs))
    dom = eigvals[dom_idx]
    return {
        'spectral_radius': float(abs_eigs[dom_idx]),
        'frobenius_norm': float(np.linalg.norm(subW, ord='fro')),
        'dominant_eigval_real': float(np.real(dom)),
        'dominant_eigval_imag': float(np.imag(dom)),
        'has_complex_eigs': bool(np.any(np.abs(np.imag(eigvals)) > 1e-9)),
    }

records = []
for triad_index, row in df_rees_72_connected_AM.iterrows():
    triad_nodes = [row[c] for c in node_cols]
    subW = A_df.loc[triad_nodes, triad_nodes].to_numpy(dtype=float)
    np.fill_diagonal(subW, 0.0)
    node_signs = get_node_signs(row)

    rees_ES, rees_node_scores = rees_es_from_submatrix(subW, node_signs)
    wES, wES_node_scores = weighted_wES_from_submatrix(subW, node_signs)

    rec = {
        'triad_index': triad_index,
        'node_1': triad_nodes[0],
        'node_2': triad_nodes[1],
        'node_3': triad_nodes[2],
        'ei_signature': ''.join('E' if s == 1 else 'I' for s in node_signs),
        'superpattern': row.get('superpattern', np.nan),
        'superpattern_name': row.get('superpattern_name', np.nan),
        'n_edges_rees_enum': row.get('n_edges', np.nan),
        'rees_ES_recomputed': rees_ES,
        'wES_weighted_Wij': wES,
        'wES_minus_rees_ES': wES - rees_ES,
        'abs_wES_minus_rees_ES': abs(wES - rees_ES),
        'rees_node_1': rees_node_scores[0],
        'rees_node_2': rees_node_scores[1],
        'rees_node_3': rees_node_scores[2],
        'wES_node_1': wES_node_scores[0],
        'wES_node_2': wES_node_scores[1],
        'wES_node_3': wES_node_scores[2],
        'positive_weight_sum': float(subW[subW > 0].sum()),
        'negative_weight_sum': float(subW[subW < 0].sum()),
        'absolute_weight_sum': float(np.abs(subW).sum()),
        'nonzero_directed_edges_Wij': int(np.count_nonzero(subW)),
    }
    rec.update(spectral_metrics(subW))
    records.append(rec)

df_rees72_wes_compare = pd.DataFrame(records)

summary_by_superpattern = (
    df_rees72_wes_compare.groupby('superpattern', dropna=False)
    .agg(
        n_trimers=('triad_index', 'count'),
        mean_rees_ES=('rees_ES_recomputed', 'mean'),
        median_rees_ES=('rees_ES_recomputed', 'median'),
        mean_wES=('wES_weighted_Wij', 'mean'),
        median_wES=('wES_weighted_Wij', 'median'),
        mean_abs_shift=('abs_wES_minus_rees_ES', 'mean'),
        median_abs_drive=('absolute_weight_sum', 'median'),
        mean_spectral_radius=('spectral_radius', 'mean'),
    ).reset_index().sort_values('superpattern')
)

summary_by_ei = (
    df_rees72_wes_compare.groupby('ei_signature', dropna=False)
    .agg(
        n_trimers=('triad_index', 'count'),
        mean_rees_ES=('rees_ES_recomputed', 'mean'),
        mean_wES=('wES_weighted_Wij', 'mean'),
        mean_abs_shift=('abs_wES_minus_rees_ES', 'mean'),
        median_abs_drive=('absolute_weight_sum', 'median'),
        mean_spectral_radius=('spectral_radius', 'mean'),
    ).reset_index().sort_values('n_trimers', ascending=False)
)

top100_largest_shift = df_rees72_wes_compare.sort_values('abs_wES_minus_rees_ES', ascending=False).head(100)

display(df_rees72_wes_compare.head())
display(summary_by_superpattern)
display(summary_by_ei)
display(top100_largest_shift.head(20))

df_rees72_wes_compare.to_csv(OUTPUT_DIR / 'section5_rees72_connected_AM_reesES_vs_weightedWij_wES.csv', index=False)
summary_by_superpattern.to_csv(OUTPUT_DIR / 'section5_summary_by_superpattern.csv', index=False)
summary_by_ei.to_csv(OUTPUT_DIR / 'section5_summary_by_ei_signature.csv', index=False)
top100_largest_shift.to_csv(OUTPUT_DIR / 'section5_top100_largest_reesES_wES_shift.csv', index=False)
```

## 6. Interactive plots

Interactive HTML plots are exported to the output directory. To avoid bloated notebooks, set `SHOW_FIGS = True` only when running interactively.

```python
# ============================================================
# 6. Interactive plots
# ============================================================

SHOW_FIGS = False

if px is None:
    print('Plotly unavailable; skipping interactive plots.')
else:
    plot_df = df_rees72_wes_compare.copy()

    fig = px.histogram(
        plot_df,
        x='wES_weighted_Wij',
        color='superpattern',
        nbins=150,
        title='Weighted W_ij wES Distribution by Superpattern — W72 Connected A–M Subset',
    )
    fig.update_layout(template='plotly_white', height=650, xaxis_title='weighted W_ij wES', yaxis_title='Trimer count', bargap=0.02)
    fig.add_annotation(x=0.99, y=0.98, xref='paper', yref='paper', text=f'N = {len(plot_df):,}', showarrow=False, bgcolor='white', bordercolor='black', borderwidth=1)
    fig.write_html(OUTPUT_DIR / 'section6_interactive_wES_by_superpattern.html')
    if SHOW_FIGS:
        fig.show()

    fig = px.scatter(
        plot_df,
        x='rees_ES_recomputed',
        y='wES_weighted_Wij',
        color='superpattern',
        hover_data=['node_1', 'node_2', 'node_3', 'ei_signature', 'absolute_weight_sum', 'spectral_radius'],
        title='Recomputed Rees ES vs Weighted W_ij wES',
    )
    fig.update_layout(template='plotly_white', height=700, xaxis_title='Recomputed Rees ES', yaxis_title='Weighted W_ij wES')
    fig.write_html(OUTPUT_DIR / 'section6_interactive_reesES_vs_wES_scatter.html')
    if SHOW_FIGS:
        fig.show()

    fig = px.bar(
        superpattern_count_audit,
        x='superpattern',
        y=['rees_enum_count', 'W72_subset_count', 'W72_connected_AM_count'],
        barmode='group',
        title='Superpattern Count Audit: Rees Enumeration vs W72 Subsets',
    )
    fig.update_layout(template='plotly_white', height=600, xaxis_title='Superpattern', yaxis_title='Trimer count')
    fig.write_html(OUTPUT_DIR / 'section6_interactive_superpattern_count_audit.html')
    if SHOW_FIGS:
        fig.show()

    print('Interactive HTML plots saved to:', OUTPUT_DIR)
```
