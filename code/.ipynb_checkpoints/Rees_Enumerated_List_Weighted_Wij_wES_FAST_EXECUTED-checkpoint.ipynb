# Rees-Enumerated Trimer List Scored with Weighted `W_ij` wES

This notebook preserves the prior Rees-style enumeration table (`stage2a_triad_enumeration_all.csv`) and appends weighted information from `w_ij_gaa.csv`.

It is optimized to score all 59,640 trimers quickly using NumPy array indexing rather than slow row-wise matrix slicing.

Outputs include:
- all enumerated trimers scored with weighted `W_ij`
- connected Rees A–M trimers scored with weighted `W_ij`
- wES, Rees ES, spectral metrics, and six directed `W_ij` values per trimer
- rankings, group summaries, and interactive HTML plots

```python
# ============================================================
# 0. Setup: Colab/Drive detection + imports + paths
# ============================================================
from pathlib import Path
import os, sys, re, warnings
warnings.filterwarnings("ignore")

IN_COLAB = "google.colab" in sys.modules

if IN_COLAB:
    from google.colab import drive
    DRIVE_MOUNT = Path("/content/drive")
    drive.mount(str(DRIVE_MOUNT))
    PROJECT_DIR = Path("/content/drive/MyDrive/Ascoli_LabRotation")
    if PROJECT_DIR.exists():
        os.chdir(PROJECT_DIR)
    else:
        PROJECT_DIR = Path.cwd()
else:
    PROJECT_DIR = Path.cwd()

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    import plotly.express as px
except Exception:
    px = None

possible_weight_paths = [
    Path("/mnt/data/w_ij_gaa.csv"),
    PROJECT_DIR / "matrices" / "w_ij_gaa.csv",
    PROJECT_DIR / "w_ij_gaa.csv",
    Path("w_ij_gaa.csv"),
]
possible_triad_paths = [
    Path("/mnt/data/stage2a_triad_enumeration_all.csv"),
    PROJECT_DIR / "outputs" / "stage2a_triad_enumeration_all.csv",
    PROJECT_DIR / "stage2a_triad_enumeration_all.csv",
    Path("stage2a_triad_enumeration_all.csv"),
]

def first_existing(paths):
    for p in paths:
        if p.exists() and p.is_file():
            return p
    return None

WEIGHT_PATH = first_existing(possible_weight_paths)
TRIAD_PATH = first_existing(possible_triad_paths)

if WEIGHT_PATH is None:
    raise FileNotFoundError("Could not locate w_ij_gaa.csv. Edit WEIGHT_PATH manually.")
if TRIAD_PATH is None:
    raise FileNotFoundError("Could not locate stage2a_triad_enumeration_all.csv. Edit TRIAD_PATH manually.")

OUTPUT_DIR = PROJECT_DIR / "outputs_rees_enumerated_weighted_wij_wES_fast"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("IN_COLAB    :", IN_COLAB)
print("PROJECT_DIR :", PROJECT_DIR)
print("WEIGHT_PATH :", WEIGHT_PATH)
print("TRIAD_PATH  :", TRIAD_PATH)
print("OUTPUT_DIR  :", OUTPUT_DIR)
```

## 1. Load the weighted matrix and Rees-style enumeration

The enumeration table is treated as authoritative. The notebook does not rebuild the motif list; it only applies weighted `W_ij` values to that list.

```python
# ============================================================
# 1. Load inputs
# ============================================================
A_df = pd.read_csv(WEIGHT_PATH, index_col=0)
A_df = A_df.apply(pd.to_numeric, errors="coerce").fillna(0.0)
A_df.index = A_df.index.astype(str).str.strip()
A_df.columns = A_df.columns.astype(str).str.strip()

if set(A_df.index) == set(A_df.columns):
    A_df = A_df.loc[A_df.index, A_df.index]

if A_df.shape[0] != A_df.shape[1]:
    raise ValueError(f"W_ij matrix is not square: {A_df.shape}")
if list(A_df.index) != list(A_df.columns):
    raise ValueError("W_ij row/column labels do not match.")

A = A_df.to_numpy(dtype=float)
labels = list(A_df.index)
node_to_idx = {node: idx for idx, node in enumerate(labels)}

df_enum = pd.read_csv(TRIAD_PATH)

print("W_ij shape:", A_df.shape)
print("Enumeration shape:", df_enum.shape)
display(df_enum.head())
```

## 2. Standardize node columns and reconcile labels

A known mismatch appears between the enumeration and the weighted matrix:

`MEC LIII Multipolar Principal` → `LEC LIII Multipolar Principal`

This is remapped before scoring so the same enumerated trimer list can be applied to `W_ij`.

```python
# ============================================================
# 2. Standardize trimer columns + reconcile labels
# ============================================================
df_triads = df_enum.copy()

if all(c in df_triads.columns for c in ["node1", "node2", "node3"]):
    df_triads = df_triads.rename(columns={"node1": "node_1", "node2": "node_2", "node3": "node_3"})
elif all(c in df_triads.columns for c in ["i", "j", "k"]):
    df_triads = df_triads.rename(columns={"i": "node_1", "j": "node_2", "k": "node_3"})
elif all(c in df_triads.columns for c in ["A", "B", "C"]):
    df_triads = df_triads.rename(columns={"A": "node_1", "B": "node_2", "C": "node_3"})
else:
    raise ValueError("Expected node1/node2/node3, i/j/k, or A/B/C columns.")

for c in ["node_1", "node_2", "node_3"]:
    df_triads[c] = df_triads[c].astype(str).str.strip()
    df_triads[c + "_original"] = df_triads[c]

manual_label_map = {
    "MEC LIII Multipolar Principal": "LEC LIII Multipolar Principal",
}

for c in ["node_1", "node_2", "node_3"]:
    df_triads[c] = df_triads[c].map(lambda x: manual_label_map.get(x, x))

all_nodes = sorted(pd.unique(df_triads[["node_1", "node_2", "node_3"]].values.ravel("K")))
missing = sorted([n for n in all_nodes if n not in node_to_idx])

print("Unique nodes in enumeration after remap:", len(all_nodes))
print("Missing after remap:", missing)
if missing:
    raise ValueError("Some labels in the trimer enumeration still do not exist in W_ij.")

label_audit = pd.DataFrame({"original_label": sorted(pd.unique(df_triads[["node_1_original", "node_2_original", "node_3_original"]].values.ravel("K")))})
label_audit["resolved_label"] = label_audit["original_label"].map(lambda x: manual_label_map.get(x, x))
label_audit["changed"] = label_audit["original_label"] != label_audit["resolved_label"]
label_audit.to_csv(OUTPUT_DIR / "section2_label_reconciliation_audit.csv", index=False)

df_triads.to_csv(OUTPUT_DIR / "section2_standardized_rees_enumerated_trimers.csv", index=False)

display(label_audit[label_audit["changed"]])
display(df_triads.head())
```

## 3. Build fast indexed 3×3 submatrices

Each row of the enumeration is converted to numeric matrix indices, then all 3×3 weighted submatrices are extracted in one NumPy operation.

```python
# ============================================================
# 3. Fast trimer submatrix extraction
# ============================================================
triad_index_array = np.array(
    [[node_to_idx[a], node_to_idx[b], node_to_idx[c]] for a, b, c in df_triads[["node_1", "node_2", "node_3"]].itertuples(index=False)],
    dtype=int,
)

subW_all = A[triad_index_array[:, :, None], triad_index_array[:, None, :]].copy()
subW_all[:, np.arange(3), np.arange(3)] = 0.0

print("Triad index array:", triad_index_array.shape)
print("Weighted trimer submatrix array:", subW_all.shape)
print("Memory MB:", round(subW_all.nbytes / 1e6, 2))
```

## 4. Compute weighted ES, Rees ES, and spectral metrics

Node-level weighted ES:

$$wES_i = s_i \left(1 + lpha \log(1 + E_i) - eta \log(1 + I_i)ight)$$

Total trimer score:

$$wES_{triad} = \sum_{i=1}^{3} wES_i$$

Spectral radius:

$$ho(W_{triad}) = \max_i |\lambda_i|$$

```python
# ============================================================
# 4. Vectorized weighted scoring
# ============================================================
ALPHA = 0.10
BETA = 0.10

# E/I signs from enumeration metadata when available.
def ei_to_sign_array(vals):
    arr = pd.Series(vals).astype(str).str.strip().str.lower()
    return np.where(arr.str.startswith("e"), 1, -1)

if all(c in df_triads.columns for c in ["ei_1", "ei_2", "ei_3"]):
    node_signs = np.column_stack([
        ei_to_sign_array(df_triads["ei_1"]),
        ei_to_sign_array(df_triads["ei_2"]),
        ei_to_sign_array(df_triads["ei_3"]),
    ]).astype(int)
else:
    # Fallback: infer by outgoing signed balance.
    out_sum = A.sum(axis=1)
    node_signs = np.where(out_sum[triad_index_array] >= 0, 1, -1).astype(int)

positive = np.where(subW_all > 0, subW_all, 0.0)
negative_abs = np.where(subW_all < 0, -subW_all, 0.0)

# Incoming drive is column-wise, so sum over source axis = axis 1.
incoming_exc = positive.sum(axis=1)
incoming_inh = negative_abs.sum(axis=1)

wES_nodes = node_signs * (1 + ALPHA * np.log1p(incoming_exc) - BETA * np.log1p(incoming_inh))
wES = wES_nodes.sum(axis=1)

# Rees-style ES: categorical multipliers based on incoming edge sign.
rees_nodes = node_signs.astype(float).copy()
for target in range(3):
    for source in range(3):
        if source == target:
            continue
        w = subW_all[:, source, target]
        rees_nodes[:, target] *= np.where(w > 0, 1.1, np.where(w < 0, 0.9, 1.0))
rees_ES = rees_nodes.sum(axis=1)

# Edge and weight summaries.
n_directed_edges_weighted = np.count_nonzero(subW_all, axis=(1, 2))
pos_weight_sum = positive.sum(axis=(1, 2))
neg_weight_sum = np.where(subW_all < 0, subW_all, 0.0).sum(axis=(1, 2))
abs_weight_sum = np.abs(subW_all).sum(axis=(1, 2))
net_weight_sum = subW_all.sum(axis=(1, 2))

# Spectral metrics.
eigvals = np.linalg.eigvals(subW_all)
abs_eig = np.abs(eigvals)
dom_idx = abs_eig.argmax(axis=1)
row_idx = np.arange(len(df_triads))
dominant_eig = eigvals[row_idx, dom_idx]
spectral_radius = abs_eig[row_idx, dom_idx]
frobenius_norm = np.linalg.norm(subW_all, axis=(1, 2))
determinant = np.linalg.det(subW_all)
complex_eig_count = (np.abs(np.imag(eigvals)) > 1e-9).sum(axis=1)
has_complex_eigs = complex_eig_count > 0

print("Computed wES and spectral metrics for:", len(df_triads), "trimers")
```

## 5. Assemble the scored table

```python
# ============================================================
# 5. Assemble output table
# ============================================================
base_cols = [
    "node_1", "node_2", "node_3",
    "node_1_original", "node_2_original", "node_3_original",
]
metadata_cols = [c for c in [
    "ei_1", "ei_2", "ei_3",
    "region_1", "region_2", "region_3",
    "ei_pattern", "superpattern", "superpattern_name",
    "is_connected", "n_edges", "total_abs_weight",
] if c in df_triads.columns]

df_wij_wes_all = df_triads[base_cols + metadata_cols].copy()
df_wij_wes_all.insert(0, "triad_index", np.arange(len(df_wij_wes_all)))
df_wij_wes_all["status"] = "ok"

df_wij_wes_all["node_1_sign"] = node_signs[:, 0]
df_wij_wes_all["node_2_sign"] = node_signs[:, 1]
df_wij_wes_all["node_3_sign"] = node_signs[:, 2]
df_wij_wes_all["ei_signature_used"] = ["".join("E" if s == 1 else "I" for s in row) for row in node_signs]

# Explicit directed W_ij values.
df_wij_wes_all["W_12"] = subW_all[:, 0, 1]
df_wij_wes_all["W_21"] = subW_all[:, 1, 0]
df_wij_wes_all["W_13"] = subW_all[:, 0, 2]
df_wij_wes_all["W_31"] = subW_all[:, 2, 0]
df_wij_wes_all["W_23"] = subW_all[:, 1, 2]
df_wij_wes_all["W_32"] = subW_all[:, 2, 1]

df_wij_wes_all["n_directed_edges_weighted"] = n_directed_edges_weighted
df_wij_wes_all["positive_weight_sum_wij"] = pos_weight_sum
df_wij_wes_all["negative_weight_sum_wij"] = neg_weight_sum
df_wij_wes_all["absolute_weight_sum_wij"] = abs_weight_sum
df_wij_wes_all["net_weight_sum_wij"] = net_weight_sum

df_wij_wes_all["incoming_exc_node_1"] = incoming_exc[:, 0]
df_wij_wes_all["incoming_exc_node_2"] = incoming_exc[:, 1]
df_wij_wes_all["incoming_exc_node_3"] = incoming_exc[:, 2]
df_wij_wes_all["incoming_inh_node_1"] = incoming_inh[:, 0]
df_wij_wes_all["incoming_inh_node_2"] = incoming_inh[:, 1]
df_wij_wes_all["incoming_inh_node_3"] = incoming_inh[:, 2]

df_wij_wes_all["wES"] = wES
df_wij_wes_all["wES_node_1"] = wES_nodes[:, 0]
df_wij_wes_all["wES_node_2"] = wES_nodes[:, 1]
df_wij_wes_all["wES_node_3"] = wES_nodes[:, 2]

df_wij_wes_all["rees_ES"] = rees_ES
df_wij_wes_all["rees_ES_node_1"] = rees_nodes[:, 0]
df_wij_wes_all["rees_ES_node_2"] = rees_nodes[:, 1]
df_wij_wes_all["rees_ES_node_3"] = rees_nodes[:, 2]
df_wij_wes_all["wES_minus_rees_ES"] = df_wij_wes_all["wES"] - df_wij_wes_all["rees_ES"]
df_wij_wes_all["abs_wES_minus_rees_ES"] = df_wij_wes_all["wES_minus_rees_ES"].abs()

df_wij_wes_all["spectral_radius"] = spectral_radius
df_wij_wes_all["frobenius_norm"] = frobenius_norm
df_wij_wes_all["determinant"] = determinant
df_wij_wes_all["dominant_eigval_real"] = np.real(dominant_eig)
df_wij_wes_all["dominant_eigval_imag"] = np.imag(dominant_eig)
df_wij_wes_all["complex_eig_count"] = complex_eig_count
df_wij_wes_all["has_complex_eigs"] = has_complex_eigs

all_path = OUTPUT_DIR / "section5_ALL_rees_enumerated_trimers_weighted_Wij_wES.csv"
df_wij_wes_all.to_csv(all_path, index=False)

print("All weighted-scored rows:", df_wij_wes_all.shape)
print("Saved:", all_path)
display(df_wij_wes_all.head())
```

## 6. Extract connected Rees A–M subset

This preserves the connected motifs from the previous enumeration and removes disconnected superpattern labels such as `-B` and `-C`.

```python
# ============================================================
# 6. Connected Rees A–M subset
# ============================================================
df_connected = df_wij_wes_all.copy()

if "is_connected" in df_connected.columns:
    df_connected = df_connected[df_connected["is_connected"].astype(bool)].copy()

if "superpattern" in df_connected.columns:
    df_connected = df_connected[~df_connected["superpattern"].astype(str).str.startswith("-")].copy()

connected_path = OUTPUT_DIR / "section6_CONNECTED_Rees_A_to_M_weighted_Wij_wES.csv"
df_connected.to_csv(connected_path, index=False)

print("All scored trimers:", len(df_wij_wes_all))
print("Connected Rees A–M trimers:", len(df_connected))
print("Saved:", connected_path)

if "superpattern" in df_connected.columns:
    display(df_connected["superpattern"].value_counts(dropna=False).rename_axis("superpattern").reset_index(name="count"))
```

## 7. Rankings and summaries

```python
# ============================================================
# 7. Rankings and summary statistics
# ============================================================
rank_base = df_connected.copy()

rankings = {
    "top100_highest_wES": rank_base.sort_values("wES", ascending=False).head(100),
    "top100_lowest_wES": rank_base.sort_values("wES", ascending=True).head(100),
    "top100_absolute_weight_sum_wij": rank_base.sort_values("absolute_weight_sum_wij", ascending=False).head(100),
    "top100_spectral_radius": rank_base.sort_values("spectral_radius", ascending=False).head(100),
    "top100_largest_wES_Rees_shift": rank_base.sort_values("abs_wES_minus_rees_ES", ascending=False).head(100),
}

for name, out_df in rankings.items():
    out_df.to_csv(OUTPUT_DIR / f"section7_{name}.csv", index=False)

summary_cols = ["wES", "rees_ES", "wES_minus_rees_ES", "absolute_weight_sum_wij", "net_weight_sum_wij", "spectral_radius", "frobenius_norm"]
summary = rank_base[summary_cols].describe().T.reset_index().rename(columns={"index": "metric"})
summary.to_csv(OUTPUT_DIR / "section7_CONNECTED_summary_statistics.csv", index=False)

display(summary)
display(rankings["top100_highest_wES"].head(20))
display(rankings["top100_spectral_radius"].head(20))
```

## 8. Group summaries by superpattern and E/I pattern

```python
# ============================================================
# 8. Group summaries
# ============================================================
if "superpattern" in rank_base.columns:
    summary_by_superpattern = (
        rank_base.groupby("superpattern")
        .agg(
            n_trimers=("triad_index", "count"),
            mean_wES=("wES", "mean"),
            median_wES=("wES", "median"),
            mean_rees_ES=("rees_ES", "mean"),
            mean_abs_weight=("absolute_weight_sum_wij", "mean"),
            median_abs_weight=("absolute_weight_sum_wij", "median"),
            mean_spectral_radius=("spectral_radius", "mean"),
            median_spectral_radius=("spectral_radius", "median"),
            complex_fraction=("has_complex_eigs", "mean"),
        )
        .reset_index()
        .sort_values("n_trimers", ascending=False)
    )
    summary_by_superpattern.to_csv(OUTPUT_DIR / "section8_summary_by_superpattern.csv", index=False)
    display(summary_by_superpattern)

if "ei_pattern" in rank_base.columns:
    summary_by_ei_pattern = (
        rank_base.groupby("ei_pattern")
        .agg(
            n_trimers=("triad_index", "count"),
            mean_wES=("wES", "mean"),
            median_wES=("wES", "median"),
            mean_rees_ES=("rees_ES", "mean"),
            mean_abs_weight=("absolute_weight_sum_wij", "mean"),
            median_abs_weight=("absolute_weight_sum_wij", "median"),
            mean_spectral_radius=("spectral_radius", "mean"),
        )
        .reset_index()
        .sort_values("n_trimers", ascending=False)
    )
    summary_by_ei_pattern.to_csv(OUTPUT_DIR / "section8_summary_by_ei_pattern.csv", index=False)
    display(summary_by_ei_pattern)
```

## 9. Interactive plots

HTML plots are written to disk. Set `SHOW_FIGS = True` for inline rendering in Colab/Jupyter.

```python
# ============================================================
# 9. Interactive plots
# ============================================================
SHOW_FIGS = False

if px is None:
    print("Plotly is unavailable; skipping interactive plots.")
else:
    plot_df = rank_base.copy()
    total_n = len(plot_df)
    specs = [
        ("wES", "Weighted Excitability Score", False),
        ("wES_minus_rees_ES", "wES - Rees ES", False),
        ("absolute_weight_sum_wij", "Absolute Weighted W_ij Sum", True),
        ("spectral_radius", "Spectral Radius", True),
    ]
    for col, label, use_log in specs:
        temp = plot_df.copy()
        x = col
        x_label = label
        if use_log:
            x = "log10_" + col
            temp[x] = np.log10(temp[col] + 1)
            x_label = f"log10({label} + 1)"
        fig = px.histogram(temp, x=x, nbins=150, title=f"{label} Distribution")
        fig.update_layout(template="plotly_white", height=650, bargap=0.02, xaxis_title=x_label, yaxis_title="Trimer Count")
        fig.add_annotation(x=0.99, y=0.98, xref="paper", yref="paper", text=f"N = {total_n:,}", showarrow=False, bgcolor="white", bordercolor="black", borderwidth=1)
        outfile = OUTPUT_DIR / f"section9_interactive_{col}_histogram.html"
        fig.write_html(outfile)
        print("Saved:", outfile)
        if SHOW_FIGS:
            fig.show()

    fig = px.scatter(
        plot_df,
        x="rees_ES",
        y="wES",
        hover_data=["node_1", "node_2", "node_3", "superpattern", "ei_pattern", "absolute_weight_sum_wij", "spectral_radius"],
        title="Rees ES vs Weighted ES using weighted W_ij",
    )
    fig.update_layout(template="plotly_white", height=700)
    fig.add_annotation(x=0.99, y=0.02, xref="paper", yref="paper", text=f"N = {total_n:,}", showarrow=False, bgcolor="white", bordercolor="black", borderwidth=1)
    outfile = OUTPUT_DIR / "section9_interactive_rees_ES_vs_wES_scatter.html"
    fig.write_html(outfile)
    print("Saved:", outfile)
    if SHOW_FIGS:
        fig.show()
```

## 10. Final exports

```python
# ============================================================
# 10. Final exports
# ============================================================
final_all = OUTPUT_DIR / "FINAL_ALL_rees_enumerated_trimers_weighted_Wij_wES.csv"
final_connected = OUTPUT_DIR / "FINAL_CONNECTED_Rees_A_to_M_weighted_Wij_wES.csv"
final_summary = OUTPUT_DIR / "FINAL_run_summary.csv"

df_wij_wes_all.to_csv(final_all, index=False)
rank_base.to_csv(final_connected, index=False)

run_summary = pd.DataFrame({
    "metric": [
        "input_enumerated_trimers",
        "all_weighted_scored_trimers",
        "connected_Rees_A_to_M_trimers",
        "mean_connected_wES",
        "median_connected_wES",
        "mean_connected_rees_ES",
        "median_connected_rees_ES",
        "mean_connected_spectral_radius",
        "median_connected_spectral_radius",
        "mean_connected_absolute_weight_sum_wij",
        "median_connected_absolute_weight_sum_wij",
    ],
    "value": [
        len(df_triads),
        len(df_wij_wes_all),
        len(rank_base),
        rank_base["wES"].mean(),
        rank_base["wES"].median(),
        rank_base["rees_ES"].mean(),
        rank_base["rees_ES"].median(),
        rank_base["spectral_radius"].mean(),
        rank_base["spectral_radius"].median(),
        rank_base["absolute_weight_sum_wij"].mean(),
        rank_base["absolute_weight_sum_wij"].median(),
    ]
})
run_summary.to_csv(final_summary, index=False)

display(run_summary)
print("Saved:", final_all)
print("Saved:", final_connected)
print("Saved:", final_summary)
```
