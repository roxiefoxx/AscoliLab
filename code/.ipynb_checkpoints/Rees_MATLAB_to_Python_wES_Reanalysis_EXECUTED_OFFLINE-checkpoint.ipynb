## 2. Core functions used for Rees-style ES, wES, topology, and spectral metrics

```python
EDGE_ORDER = [(0,1), (1,0), (0,2), (2,0), (1,2), (2,1)]

def triad_code(B):
    return int(sum((1 if B[i, j] else 0) << idx for idx, (i, j) in enumerate(EDGE_ORDER)))

def canonical(B):
    return min(triad_code(B[np.ix_(p, p)]) for p in permutations(range(3)))

def directed_edge_count(B):
    return int(np.count_nonzero(B))

def nodes_touching_edge(B):
    return int(np.sum(B.any(axis=0) | B.any(axis=1)))

def is_connected_rees_A_to_M_candidate(B):
    return directed_edge_count(B) >= 2 and nodes_touching_edge(B) == 3

def disconnected_class(B):
    e = directed_edge_count(B)
    touch = nodes_touching_edge(B)
    if touch == 3:
        return 'connected_candidate'
    if e == 0:
        return 'empty'
    if e == 1:
        return '-B_single_edge_disconnected'
    if e == 2:
        for a, b in [(0,1), (0,2), (1,2)]:
            if B[a,b] and B[b,a] and touch == 2:
                return '-A_reciprocal_dyad_disconnected'
        return '-C_two_edge_disconnected'
    return 'other_disconnected'

def rees_es(subW, signs):
    scores = []
    for target in range(3):
        score = float(signs[target])
        for source in range(3):
            if source == target:
                continue
            w = subW[source, target]
            if w > 0:
                score *= 1.1
            elif w < 0:
                score *= 0.9
        scores.append(score)
    return float(np.sum(scores))

def wes(subW, signs, alpha=0.10, beta=0.10):
    scores = []
    for target in range(3):
        exc_drive = 0.0
        inh_drive = 0.0
        for source in range(3):
            if source == target:
                continue
            w = subW[source, target]
            if signs[source] == 1 and w != 0:
                exc_drive += abs(w)
            elif signs[source] == -1 and w != 0:
                inh_drive += abs(w)
        scores.append(float(signs[target]) * (1 + alpha*np.log1p(exc_drive) - beta*np.log1p(inh_drive)))
    return float(np.sum(scores))
```

## 3. 72-node weighted matrix load and summary

```python
Wdf = pd.read_csv(USER_WEIGHT_PATH, index_col=0).apply(pd.to_numeric, errors='coerce').fillna(0.0)
Wdf.index = Wdf.index.astype(str).str.strip()
Wdf.columns = Wdf.columns.astype(str).str.strip()
Wdf = Wdf.loc[Wdf.index, Wdf.index]
W = Wdf.to_numpy(float)
np.fill_diagonal(W, 0)
labels = list(Wdf.index)
signs = np.where(W.sum(axis=1) >= 0, 1, -1)

print('72-node matrix shape:', W.shape)
print('All unordered trimers:', math.comb(len(labels), 3))
print('Nonzero off-diagonal directed edges:', int(np.count_nonzero(W)))
```
