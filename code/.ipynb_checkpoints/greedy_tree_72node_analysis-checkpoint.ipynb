# Greedy Tree / Greedy Path Analysis for the 72-Node Hippocampal Connectivity Matrix

This notebook extracts the greedy path/tree workflow into a clean, self-contained analysis using the `72_wij` sheet from `w_ij_gaa.xlsx`.

Core idea: for each start node, follow the strongest eligible outgoing edge. If the strongest edge would revisit an already visited node, the rerouting version chooses the next strongest unvisited outgoing edge instead of stopping immediately.

Outputs are written to `greedy_tree_outputs/` as CSV and PNG files.

```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx

DATA_PATH = Path('w_ij_gaa.xlsx')
if not DATA_PATH.exists():
    DATA_PATH = Path('/mnt/data/w_ij_gaa.xlsx')

SHEET = '72_wij'
OUTDIR = Path('greedy_tree_outputs')
OUTDIR.mkdir(exist_ok=True)

pd.set_option('display.max_colwidth', 160)
pd.set_option('display.max_rows', 100)
```

## 1. Load and validate the 72-node signed weighted matrix

Rows are interpreted as source / presynaptic nodes and columns as target / postsynaptic nodes. Positive weights are treated as excitatory-strength edges and negative weights as inhibitory-strength edges.

```python
W_df = pd.read_excel(DATA_PATH, sheet_name=SHEET, index_col=0)
W_df = W_df.apply(pd.to_numeric, errors='coerce').fillna(0.0)

# Align row and column order if labels match but are not ordered identically.
if set(W_df.index) == set(W_df.columns):
    W_df = W_df.loc[W_df.index, W_df.index]
else:
    raise ValueError('Row and column labels do not contain the same 72 nodes.')

# Remove self-connections for path tracing unless explicitly changed later.
np.fill_diagonal(W_df.values, 0.0)

labels = list(W_df.index)
W = W_df.to_numpy(dtype=float)
N = len(labels)

summary = pd.DataFrame({
    'metric': ['nodes', 'nonzero_edges', 'positive_edges', 'negative_edges', 'min_weight', 'max_weight'],
    'value': [N, int((W != 0).sum()), int((W > 0).sum()), int((W < 0).sum()), float(W.min()), float(W.max())]
})
summary
```

## 2. Classify neuron labels as excitatory or inhibitory

The classifier uses conservative label rules. Principal/pyramidal/granule/stellate/mossy-like labels are marked excitatory unless they contain interneuron-like terms. Most named interneuron classes are marked inhibitory.

Review the table below. If any label is wrong, edit the `manual_overrides` dictionary and rerun from this section downward.

```python
inhibitory_keywords = [
    'basket', 'axo axonic', 'axo-axonic', 'bistratified', 'trilaminar', 'ivy',
    'neurogliaform', 'olm', 'o lm', 'or lm', 'o lmr', 'lm r', 'lmr', 'hipp',
    'hicap', 'hiprom', 'mopp', 'molax', 'interneuron', 'sp sr', 'oriens',
    'perforant path associated', 'schaffer collateral associated', 'lucidum',
    'mossy fiber associated', 'lm target', 'r receiving apical', 'so so', 'orden'
]

excitatory_keywords = [
    'granule', 'semilunar granule', 'mossy', 'pyramidal', 'principal',
    'stellate', 'fan', 'multiform', 'tripolar', 'bipolar', 'multipolar'
]

manual_overrides = {
    # Add corrections here, e.g. 'DG Mossy': 'E' or 'CA1 Basket': 'I'
}

def classify_node(name: str) -> str:
    if name in manual_overrides:
        return manual_overrides[name]
    s = name.lower().replace('+', ' plus ')
    if any(k in s for k in inhibitory_keywords):
        return 'I'
    if any(k in s for k in excitatory_keywords):
        return 'E'
    return 'Unknown'

node_type = {name: classify_node(name) for name in labels}
node_df = pd.DataFrame({'node': labels, 'type': [node_type[x] for x in labels]})
node_df['region'] = node_df['node'].str.extract(r'^(DG|CA1|CA2|CA3|EC|MEC|LEC)', expand=False).fillna('Other')

node_df.to_csv(OUTDIR / 'node_classification.csv', index=False)
print(node_df['type'].value_counts(dropna=False))
node_df
```

## 3. Build edge table and basic signed-weight summaries

```python
edge_records = []
for i, src in enumerate(labels):
    for j, dst in enumerate(labels):
        w = W[i, j]
        if w != 0:
            edge_records.append({
                'source': src, 'target': dst, 'weight': float(w),
                'abs_weight': float(abs(w)),
                'sign': 'positive' if w > 0 else 'negative',
                'source_type': node_type[src], 'target_type': node_type[dst]
            })

edges_df = pd.DataFrame(edge_records).sort_values('abs_weight', ascending=False)
edges_df.to_csv(OUTDIR / 'all_nonzero_edges_ranked.csv', index=False)

edge_summary = edges_df.groupby(['source_type','target_type','sign']).agg(
    n_edges=('weight','size'),
    mean_weight=('weight','mean'),
    median_weight=('weight','median'),
    max_abs_weight=('abs_weight','max'),
    total_abs_weight=('abs_weight','sum')
).reset_index().sort_values(['source_type','target_type','sign'])

edge_summary.to_csv(OUTDIR / 'edge_summary_by_type.csv', index=False)
edge_summary
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(edges_df.loc[edges_df['weight'] > 0, 'weight'], bins=40, alpha=0.7, label='positive')
ax.hist(edges_df.loc[edges_df['weight'] < 0, 'weight'], bins=40, alpha=0.7, label='negative')
ax.set_title('Signed edge-weight distribution')
ax.set_xlabel('weight')
ax.set_ylabel('edge count')
ax.legend()
fig.tight_layout()
fig.savefig(OUTDIR / 'edge_weight_distribution.png', dpi=200)
plt.show()
```

## 4. Greedy path functions

Two path modes are implemented:

`strict_stop`: choose the strongest eligible outgoing edge; stop if it would revisit a node.

`reroute_unvisited`: choose the strongest eligible outgoing edge that has not already appeared in the path. This is the cleaned version of the earlier “do not stop at a revisit; choose the next best path” logic.

```python
def candidate_targets(current, visited, mode='positive', target_type=None, exclude_self=True):
    row = W_df.loc[current].copy()
    if exclude_self and current in row.index:
        row.loc[current] = 0.0

    if mode == 'positive':
        row = row[row > 0].sort_values(ascending=False)
    elif mode == 'negative':
        row = row[row < 0].reindex(row[row < 0].abs().sort_values(ascending=False).index)
    elif mode == 'absolute':
        row = row[row != 0].reindex(row[row != 0].abs().sort_values(ascending=False).index)
    else:
        raise ValueError("mode must be 'positive', 'negative', or 'absolute'")

    if target_type in {'E', 'I'}:
        row = row[[t for t in row.index if node_type.get(t) == target_type]]
    return row


def greedy_path(start, mode='positive', start_type=None, target_type=None, max_steps=None, policy='reroute_unvisited'):
    if start_type in {'E', 'I'} and node_type.get(start) != start_type:
        return None
    if max_steps is None:
        max_steps = N - 1

    path = [start]
    edges = []
    stop_reason = 'max_steps'
    current = start

    for step in range(max_steps):
        visited = set(path)
        candidates = candidate_targets(current, visited, mode=mode, target_type=target_type)
        if candidates.empty:
            stop_reason = 'no_eligible_outgoing_edge'
            break

        if policy == 'strict_stop':
            nxt = candidates.index[0]
            if nxt in visited:
                stop_reason = 'top_choice_revisit'
                break
            chosen_weight = float(candidates.iloc[0])
        elif policy == 'reroute_unvisited':
            unvisited = candidates[[t not in visited for t in candidates.index]]
            if unvisited.empty:
                stop_reason = 'all_eligible_targets_visited'
                break
            nxt = unvisited.index[0]
            chosen_weight = float(unvisited.iloc[0])
        else:
            raise ValueError("policy must be 'strict_stop' or 'reroute_unvisited'")

        edges.append((current, nxt, chosen_weight))
        path.append(nxt)
        current = nxt
    else:
        stop_reason = 'max_steps_reached'

    return {
        'start': start,
        'terminus': path[-1],
        'path': path,
        'path_string': ' → '.join(path),
        'path_length_edges': len(edges),
        'total_weight': float(sum(w for _, _, w in edges)),
        'total_abs_weight': float(sum(abs(w) for _, _, w in edges)),
        'mean_abs_weight': float(np.mean([abs(w) for _, _, w in edges])) if edges else 0.0,
        'min_weight': float(min([w for _, _, w in edges])) if edges else 0.0,
        'max_weight': float(max([w for _, _, w in edges])) if edges else 0.0,
        'stop_reason': stop_reason,
        'edges': edges,
    }


def run_greedy_set(starts, mode='positive', target_type=None, policy='reroute_unvisited', label='run'):
    records = []
    edge_rows = []
    for start in starts:
        result = greedy_path(start, mode=mode, target_type=target_type, policy=policy)
        if result is None:
            continue
        row = {k: v for k, v in result.items() if k not in {'path', 'edges'}}
        row['n_unique_nodes'] = len(set(result['path']))
        row['mode'] = mode
        row['target_type_filter'] = target_type if target_type else 'any'
        row['policy'] = policy
        row['run_label'] = label
        records.append(row)
        for step, (src, dst, w) in enumerate(result['edges'], start=1):
            edge_rows.append({
                'run_label': label, 'start': start, 'step': step,
                'source': src, 'target': dst, 'weight': w, 'abs_weight': abs(w),
                'source_type': node_type[src], 'target_type': node_type[dst]
            })
    return pd.DataFrame(records), pd.DataFrame(edge_rows)
```

## 5. Run cleaned greedy analyses

The default analysis runs three interpretable versions:

1. `E_only_positive`: start from excitatory nodes, follow only positive edges to excitatory targets.
2. `all_nodes_absolute`: start from every node, follow the largest absolute outgoing edge, regardless of sign.
3. `E_start_signed_abs`: start from excitatory nodes, follow the largest absolute outgoing edge, allowing inhibitory transitions.

```python
E_nodes = node_df.loc[node_df['type'] == 'E', 'node'].tolist()
I_nodes = node_df.loc[node_df['type'] == 'I', 'node'].tolist()

runs = []
edge_runs = []

configs = [
    {'label': 'E_only_positive', 'starts': E_nodes, 'mode': 'positive', 'target_type': 'E', 'policy': 'reroute_unvisited'},
    {'label': 'all_nodes_absolute', 'starts': labels, 'mode': 'absolute', 'target_type': None, 'policy': 'reroute_unvisited'},
    {'label': 'E_start_signed_abs', 'starts': E_nodes, 'mode': 'absolute', 'target_type': None, 'policy': 'reroute_unvisited'},
]

for cfg in configs:
    df_paths, df_path_edges = run_greedy_set(**cfg)
    runs.append(df_paths)
    edge_runs.append(df_path_edges)

paths_df = pd.concat(runs, ignore_index=True)
path_edges_df = pd.concat(edge_runs, ignore_index=True)

paths_df.to_csv(OUTDIR / 'greedy_paths_summary.csv', index=False)
path_edges_df.to_csv(OUTDIR / 'greedy_path_edges_long.csv', index=False)

paths_df.sort_values(['run_label','path_length_edges','total_abs_weight'], ascending=[True, False, False]).head(20)
```

## 6. Results summary tables

```python
run_summary = paths_df.groupby('run_label').agg(
    n_starts=('start','size'),
    mean_path_length=('path_length_edges','mean'),
    median_path_length=('path_length_edges','median'),
    max_path_length=('path_length_edges','max'),
    mean_total_abs_weight=('total_abs_weight','mean'),
    max_total_abs_weight=('total_abs_weight','max'),
    n_unique_termini=('terminus','nunique')
).reset_index()
run_summary.to_csv(OUTDIR / 'run_summary.csv', index=False)
run_summary
```

```python
terminus_counts = (paths_df.groupby(['run_label','terminus'])
                   .size().reset_index(name='n_starts')
                   .sort_values(['run_label','n_starts'], ascending=[True, False]))
terminus_counts.to_csv(OUTDIR / 'terminus_counts.csv', index=False)
terminus_counts.groupby('run_label').head(10)
```

```python
visit_records = []
for _, row in paths_df.iterrows():
    nodes = row['path_string'].split(' → ')
    for order, node in enumerate(nodes):
        visit_records.append({'run_label': row['run_label'], 'start': row['start'], 'node': node, 'order': order})
visit_df = pd.DataFrame(visit_records)
visit_counts = (visit_df.groupby(['run_label','node']).size().reset_index(name='visit_count')
                .merge(node_df, on='node', how='left')
                .sort_values(['run_label','visit_count'], ascending=[True, False]))
visit_counts.to_csv(OUTDIR / 'visit_counts.csv', index=False)
visit_counts.groupby('run_label').head(15)
```

## 8. Greedy transition graph

This graph collapses the selected transitions across all starts for a selected run. Wider edges indicate that the transition was reused by more start nodes.

```python
SELECTED_RUN = 'E_only_positive'  # Change to 'all_nodes_absolute' or 'E_start_signed_abs'
selected_edges = path_edges_df[path_edges_df['run_label'] == SELECTED_RUN].copy()
transition_counts = (selected_edges.groupby(['source','target'])
                     .agg(n=('start','nunique'), mean_abs_weight=('abs_weight','mean'))
                     .reset_index()
                     .sort_values('n', ascending=False))
transition_counts.to_csv(OUTDIR / f'transition_counts_{SELECTED_RUN}.csv', index=False)

G = nx.DiGraph()
for _, row in node_df.iterrows():
    G.add_node(row['node'], ntype=row['type'], region=row['region'])
for _, row in transition_counts.iterrows():
    G.add_edge(row['source'], row['target'], n=row['n'], weight=row['mean_abs_weight'])

# Keep the graph readable: plot transitions reused by at least two starts, or top 35 transitions.
plot_edges = transition_counts[(transition_counts['n'] >= 2)].head(35)
if plot_edges.empty:
    plot_edges = transition_counts.head(35)
H = nx.DiGraph()
for _, r in plot_edges.iterrows():
    H.add_edge(r['source'], r['target'], n=r['n'])

fig, ax = plt.subplots(figsize=(12, 8))
pos = nx.spring_layout(H, seed=7, k=1.2)
node_sizes = [300 + 80 * visit_counts[(visit_counts['run_label'] == SELECTED_RUN) & (visit_counts['node'] == n)]['visit_count'].sum() for n in H.nodes]
edge_widths = [max(1, H[u][v]['n']) for u, v in H.edges]
nx.draw_networkx_nodes(H, pos, node_size=node_sizes, ax=ax)
nx.draw_networkx_edges(H, pos, width=edge_widths, arrows=True, arrowsize=15, ax=ax)
nx.draw_networkx_labels(H, pos, font_size=7, ax=ax)
ax.set_title(f'Greedy transition graph: {SELECTED_RUN}')
ax.axis('off')
fig.tight_layout()
fig.savefig(OUTDIR / f'greedy_transition_graph_{SELECTED_RUN}.png', dpi=200)
plt.show()
```

## 9. Vertical path plot for the longest path

This is useful when the path is too deep for a standard network layout.

```python
def plot_vertical_path(path_row, outfile=None):
    nodes = path_row['path_string'].split(' → ')
    y = list(range(len(nodes)))[::-1]
    x = [0] * len(nodes)
    fig_height = max(5, 0.35 * len(nodes))
    fig, ax = plt.subplots(figsize=(8, fig_height))
    ax.scatter(x, y, s=90)
    for idx, node in enumerate(nodes):
        ax.text(0.05, y[idx], f'{idx:02d}. {node} [{node_type.get(node, "?")}]', va='center', fontsize=8)
    for idx in range(len(nodes)-1):
        ax.annotate('', xy=(0, y[idx+1]+0.12), xytext=(0, y[idx]-0.12), arrowprops=dict(arrowstyle='->', lw=1))
    ax.set_ylim(min(y)-1, max(y)+1)
    ax.set_xlim(-0.2, 1.0)
    ax.set_title(f"Longest path in {path_row['run_label']} from {path_row['start']}\nlength={path_row['path_length_edges']}, terminus={path_row['terminus']}")
    ax.axis('off')
    fig.tight_layout()
    if outfile:
        fig.savefig(outfile, dpi=200, bbox_inches='tight')
    plt.show()

longest = paths_df.sort_values(['path_length_edges','total_abs_weight'], ascending=[False, False]).iloc[0]
plot_vertical_path(longest, OUTDIR / 'longest_vertical_path.png')
longest[['run_label','start','terminus','path_length_edges','total_abs_weight','stop_reason','path_string']]
```
