# Stage 0: Excitatory Scaffold
## Hippocampal Connectome Analysis | Ascoli Lab, George Mason University

Constructs the dominant information-flow skeleton of the 72-node hippocampal connectome
by isolating the 27-node excitatory subnetwork and tracing greedy paths through it.

**Stopping rule:** At each step, follow the highest-weight outgoing edge to an unvisited node.
Stop when no unvisited node has an outgoing edge from the current node (all outgoing neighbors
have been visited). This is the exhaustion rule — no revisits, no arbitrary caps.

**Analyses:**
1. Full 72-node EDA — weight matrix structure, E/I asymmetry
2. Excitatory submatrix and graph construction
3. Hub rankings — out-strength, in-strength, overall
4. PageRank and Markov stationary distribution
5. Greedy path tracing — all 28 excitatory starting nodes
6. Terminus diagnosis — attractor pair identification
7. Short cycle enumeration — reciprocal pairs, triangles, squares
8. Figures — weight matrix, hub rankings, centrality, greedy tree, region flow, path detail

**Matrix convention:** `W[row, col]` = weight FROM row TO col (pre→post).
Row sum = total outgoing weight from a node.

**References:** Hippocampome.org v1.0; Rees et al. (eNeuro 2016)

## 2. Load Weight Matrix & Metadata

Load `w_ij_gaa.csv` and `izhikevich_72_v3.csv`. Apply MEC naming patch at load time.
Remove self-loops from the working matrix (diagonal zeroed; self-connections are noted but
excluded from path analysis).

```python
W_raw = pd.read_csv(WEIGHT_MATRIX, index_col=0)
W_raw.index   = W_raw.index.str.strip()
W_raw.columns = W_raw.columns.str.strip()
W_raw.index   = [MEC_PATCH.get(n, n) for n in W_raw.index]
W_raw.columns = [MEC_PATCH.get(n, n) for n in W_raw.columns]

ei_df = pd.read_csv(NEURON_PARAMS)
ei_df['Neuron Type'] = ei_df['Neuron Type'].str.strip().replace(MEC_PATCH)
ei_df['E/I'] = ei_df['E/I'].str.strip().str.lower()
ei_map = dict(zip(ei_df['Neuron Type'], ei_df['E/I']))

all_nodes  = list(W_raw.index)
excitatory = [n for n in all_nodes if ei_map.get(n, 'i') == 'e']
inhibitory = [n for n in all_nodes if ei_map.get(n, 'i') == 'i']

# Remove self-loops for analysis (preserve raw matrix separately)
W = W_raw.copy()
for nd in W.index: W.loc[nd, nd] = 0.0
n_selfloops = int((np.diag(W_raw.values) != 0).sum())

print(f'Matrix shape      : {W_raw.shape}')
print(f'Self-loops in raw : {n_selfloops} (zeroed in working matrix)')
print(f'Excitatory nodes  : {len(excitatory)}')
print(f'Inhibitory nodes  : {len(inhibitory)}')
print(f'Non-zero edges    : {int((W.values != 0).sum())}')
```

## 3. Full-Matrix EDA

Row sums (outgoing) and column sums (incoming) for all 72 nodes.
E neurons are net sources (positive row sums); I neurons are net sinks (negative row sums)
because inhibitory outgoing weights are negative in the signed matrix.

```python
row_sums = W.sum(axis=1)
col_sums = W.sum(axis=0)

eda_df = pd.DataFrame({
    'region'      : [region_of(n) for n in all_nodes],
    'ei'          : [ei_map.get(n, '?') for n in all_nodes],
    'out_strength': row_sums.values,
    'in_strength' : col_sums.values,
    'net_strength': (row_sums + col_sums).values,
}, index=all_nodes)
eda_df.index.name = 'node'

e_pos = (eda_df.loc[excitatory, 'net_strength'] > 0).sum()
i_neg = (eda_df.loc[inhibitory, 'net_strength'] < 0).sum()
print(f'E nodes with net positive strength: {e_pos}/{len(excitatory)}')
print(f'I nodes with net negative strength: {i_neg}/{len(inhibitory)}')
print(f'\nTop 5 out-strength (all 72 nodes):')
print(eda_df.nlargest(5, 'out_strength')[['region','ei','out_strength']].to_string())
print(f'\nBottom 5 out-strength:')
print(eda_df.nsmallest(5, 'out_strength')[['region','ei','out_strength']].to_string())
eda_df.to_csv(f'{TABLE_DIR}stage0_eda_fullmatrix.csv')
```

## 4. Build Excitatory Submatrix & Graph

Extract the 28-node E→E submatrix. Clip to zero (no negative E→E weights expected).
Remove self-loops from the path graph. Build a weighted directed graph.

```python
W_exc = W.loc[excitatory, excitatory].copy().clip(lower=0)
W_exc_noself = W_exc.copy()
for nd in W_exc_noself.index: W_exc_noself.loc[nd, nd] = 0.0

G = nx.DiGraph()
G.add_nodes_from(excitatory)
for pre in excitatory:
    for post in excitatory:
        if pre == post: continue
        w = W_exc_noself.loc[pre, post]
        if w > 0: G.add_edge(pre, post, weight=float(w))

N  = len(excitatory)
NE = G.number_of_edges()
n_possible = N * (N - 1)
print(f'Excitatory scaffold')
print(f'  Nodes       : {N}')
print(f'  Edges       : {NE} / {n_possible} possible  ({NE/n_possible*100:.1f}% density)')
print(f'  Strongly connected components: {nx.number_strongly_connected_components(G)}')
print(f'  Avg out-degree: {NE/N:.1f}')
```

```python
# Order nodes by region for readability
region_order = ['DG','CA3','CA2','CA1','MEC','LEC','EC']
ordered_E = []
for r in region_order:
    ordered_E += [n for n in excitatory if region_of(n) == r]

W_ord = W_exc_noself.loc[ordered_E, ordered_E]
arr   = W_ord.values.copy()
arr_plot = np.where(arr > 0, arr, np.nan)

fig, ax = plt.subplots(figsize=(13, 11))
im = ax.imshow(arr_plot, cmap='YlOrRd',
               norm=LogNorm(vmin=np.nanmin(arr_plot[~np.isnan(arr_plot)]),
                            vmax=np.nanmax(arr_plot)), aspect='auto')
plt.colorbar(im, ax=ax, label='Synaptic weight (log scale)', fraction=0.03, pad=0.01)

short_labels = [n.replace('Pyramidal','Pyr').replace('Principal','Princ')
                 .replace('Multipolar','Multi').replace('Complex','Cplx')[:32]
                for n in ordered_E]
ax.set_xticks(range(len(ordered_E))); ax.set_xticklabels(short_labels, rotation=90, fontsize=6.5)
ax.set_yticks(range(len(ordered_E))); ax.set_yticklabels(short_labels, fontsize=6.5)

for tick, nd in zip(ax.get_xticklabels(), ordered_E):
    tick.set_color(REGION_COLORS.get(region_of(nd), '#7F8C8D'))
for tick, nd in zip(ax.get_yticklabels(), ordered_E):
    tick.set_color(REGION_COLORS.get(region_of(nd), '#7F8C8D'))

# Region boundary lines
boundaries, prev_r = [], None
for i, nd in enumerate(ordered_E):
    r = region_of(nd)
    if r != prev_r:
        if i > 0: boundaries.append(i - 0.5)
        prev_r = r
for b in boundaries:
    ax.axhline(b, color='white', linewidth=1.2, alpha=0.8)
    ax.axvline(b, color='white', linewidth=1.2, alpha=0.8)

ax.set_title('Stage 0: Excitatory Scaffold Weight Matrix (28×28 E→E)\n'
             'Rows = presynaptic  |  Columns = postsynaptic  |  Zero = white',
             fontsize=11, fontweight='bold')
ax.set_xlabel('Postsynaptic (target)', fontsize=10)
ax.set_ylabel('Presynaptic (source)', fontsize=10)
legend_patches = [mpatches.Patch(color=REGION_COLORS[r], label=r) for r in region_order]
ax.legend(handles=legend_patches, loc='upper right', fontsize=8,
          bbox_to_anchor=(1.18, 1.0), title='Region')
plt.tight_layout()
fig.savefig(f'{TABLE_DIR}fig1_weight_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 5. Hub Rankings

```python
out_str = {n: sum(d['weight'] for _,_,d in G.out_edges(n, data=True)) for n in excitatory}
in_str  = {n: sum(d['weight'] for _,_,d in G.in_edges(n,  data=True)) for n in excitatory}

hub_df = pd.DataFrame({
    'region'          : [region_of(n) for n in excitatory],
    'out_strength'    : [out_str[n] for n in excitatory],
    'in_strength'     : [in_str[n]  for n in excitatory],
    'overall_strength': [out_str[n]+in_str[n] for n in excitatory],
    'out_degree'      : [G.out_degree(n) for n in excitatory],
    'in_degree'       : [G.in_degree(n)  for n in excitatory],
}, index=excitatory)
hub_df.index.name = 'node'
hub_df['out_rank']     = hub_df['out_strength'].rank(ascending=False).astype(int)
hub_df['in_rank']      = hub_df['in_strength'].rank(ascending=False).astype(int)
hub_df['overall_rank'] = hub_df['overall_strength'].rank(ascending=False).astype(int)
hub_df['rank_delta']   = hub_df['out_rank'] - hub_df['in_rank']
hub_df.to_csv(f'{TABLE_DIR}stage0_hub_rankings.csv')

print('Top 5 by out-strength:')
print(hub_df.nlargest(5,'out_strength')[['region','out_strength','out_rank']].to_string())
print('\nTop 5 by in-strength:')
print(hub_df.nlargest(5,'in_strength')[['region','in_strength','in_rank']].to_string())
```

## 6. PageRank & Markov Stationary Distribution

Two diffusion-based centrality measures on the E→E subgraph.

**PageRank** (α=0.85): random walk with 15% teleportation. Prevents absorption at structural sinks.

**Markov stationary distribution:** Row-normalise W_exc to form a transition matrix P.
Power-iterate `π = πP` to convergence. Represents long-run fraction of time a random
walker spends at each node.

Both differ from the greedy terminus: random walkers teleport away from dead-ends,
whereas the greedy walk is deterministically trapped there. The disagreement is
scientifically informative — the network routes *deterministic* signal toward the
greedy terminus but distributes *stochastic* activity through high-degree hubs.

```python
pr = nx.pagerank(G, weight='weight', alpha=0.85)

P = W_exc_noself.copy()
rs = P.sum(axis=1); rs[rs == 0] = 1.0
P  = P.div(rs, axis=0)
pi = np.ones(N) / N
for _ in range(10000):
    pi_new = pi @ P.values
    if np.max(np.abs(pi_new - pi)) < 1e-12: break
    pi = pi_new
markov = dict(zip(excitatory, pi))

centrality_df = pd.DataFrame({
    'region'  : [region_of(n) for n in excitatory],
    'pagerank': [pr[n] for n in excitatory],
    'markov'  : [markov[n] for n in excitatory],
}, index=excitatory)
centrality_df.index.name = 'node'
centrality_df.to_csv(f'{TABLE_DIR}stage0_centrality.csv')

print('Top 5 PageRank:')
print(centrality_df.nlargest(5,'pagerank')[['region','pagerank']].to_string())
print('\nTop 5 Markov stationary:')
print(centrality_df.nlargest(5,'markov')[['region','markov']].to_string())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 8))

# PageRank
df_p = centrality_df.reset_index().sort_values('pagerank', ascending=True)
colors = [REGION_COLORS.get(region_of(n),'#7F8C8D') for n in df_p['node']]
axes[0].barh(range(len(df_p)), df_p['pagerank'], color=colors, edgecolor='white', linewidth=0.4)
axes[0].set_yticks(range(len(df_p))); axes[0].set_yticklabels([n[:34] for n in df_p['node']], fontsize=9)
axes[0].set_xlabel('PageRank score', fontsize=9)
axes[0].set_title('PageRank (α=0.85)', fontsize=10, fontweight='bold')
for ytick, nd in zip(axes[0].get_yticklabels(), df_p['node']):
    ytick.set_color(REGION_COLORS.get(region_of(nd),'#7F8C8D'))

# Markov
df_m = centrality_df.reset_index().sort_values('markov', ascending=True)
colors_m = [REGION_COLORS.get(region_of(n),'#7F8C8D') for n in df_m['node']]
axes[1].barh(range(len(df_m)), df_m['markov'], color=colors_m, edgecolor='white', linewidth=0.4)
axes[1].set_yticks(range(len(df_m))); axes[1].set_yticklabels([n[:34] for n in df_m['node']], fontsize=9)
axes[1].set_xlabel('Stationary probability', fontsize=9)
axes[1].set_title('Markov Stationary Distribution', fontsize=10, fontweight='bold')
for ytick, nd in zip(axes[1].get_yticklabels(), df_m['node']):
    ytick.set_color(REGION_COLORS.get(region_of(nd),'#7F8C8D'))

# Out-strength
df_o = hub_df.reset_index().sort_values('out_strength', ascending=True)
colors_o = [REGION_COLORS.get(region_of(n),'#7F8C8D') for n in df_o['node']]
axes[2].barh(range(len(df_o)), np.log10(df_o['out_strength']+1), color=colors_o,
             edgecolor='white', linewidth=0.4)
axes[2].set_yticks(range(len(df_o))); axes[2].set_yticklabels([n[:34] for n in df_o['node']], fontsize=9)
axes[2].set_xlabel('log₁₀(out-strength + 1)', fontsize=9)
axes[2].set_title('Out-Strength (raw weight)', fontsize=10, fontweight='bold')
for ytick, nd in zip(axes[2].get_yticklabels(), df_o['node']):
    ytick.set_color(REGION_COLORS.get(region_of(nd),'#7F8C8D'))

for ax in axes: ax.grid(axis='x', alpha=0.25); ax.set_axisbelow(True)
legend_patches = [mpatches.Patch(color=REGION_COLORS[r], label=r)
                  for r in ['DG','CA3','CA2','CA1','MEC','LEC','EC']]
axes[2].legend(handles=legend_patches, fontsize=8, loc='lower right', title='Region')
plt.suptitle('Stage 0: Node Centrality — PageRank | Markov | Out-Strength',
             fontsize=12, fontweight='bold')
plt.tight_layout()
fig.savefig(f'{TABLE_DIR}fig3_centrality.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 7. Greedy Path Tracing

**Stopping rule:** follow the highest-weight outgoing edge to an unvisited node at each step.
Stop when no unvisited node has an outgoing edge from the current node.

**Diagnosis at terminus:** record the highest-weight outgoing target that would have been
selected next if revisits were allowed (`would_go_to`). This reveals the attempted cycle
target — the node the signal wants to return to, establishing the dominant attractor pair.

```python
def greedy_path(G, start):
    """
    Greedy path from `start` on the E→E scaffold graph.
    Stopping rule: halt when all outgoing neighbors of the current node
    have been visited (no unvisited neighbors remain).
    Returns: path, edge_weights, stop_reason, would_go_to
    """
    path, weights, visited = [start], [], {start}
    current = start
    while True:
        out = [(v, d['weight']) for _, v, d in G.out_edges(current, data=True)]
        unvisited = [(v, w) for v, w in out if v not in visited]
        if not unvisited:
            if not out:
                stop    = 'dead_end_no_outgoing'
                wgt     = None
            else:
                stop    = 'exhausted_unvisited_neighbors'
                wgt     = max(out, key=lambda x: x[1])[0]
            return path, weights, stop, wgt
        best_v, best_w = max(unvisited, key=lambda x: x[1])
        path.append(best_v); weights.append(best_w)
        visited.add(best_v); current = best_v


all_paths = []
for start in excitatory:
    path, weights, stop, would_go = greedy_path(G, start)
    all_paths.append({
        'start'          : start,
        'start_region'   : region_of(start),
        'path'           : path,
        'n_steps'        : len(path) - 1,
        'terminus'       : path[-1],
        'terminus_region': region_of(path[-1]),
        'would_go_to'    : would_go,
        'stop_reason'    : stop,
        'total_weight'   : sum(weights),
        'step_weights'   : weights,
        'path_str'       : ' -> '.join(path),
    })

paths_df = pd.DataFrame(all_paths)
paths_df[['start','start_region','n_steps','terminus','terminus_region',
          'would_go_to','stop_reason','total_weight','path_str']
].to_csv(f'{TABLE_DIR}stage0_greedy_paths.csv', index=False)

print(f'Paths traced    : {len(paths_df)}')
print(f'All stop reason : {paths_df["stop_reason"].unique()}')
print(f'Path lengths    : {paths_df["n_steps"].min()}–{paths_df["n_steps"].max()} steps')
print(f'\nTerminus distribution:')
print(paths_df["terminus"].value_counts().to_string())
print(f'\nWould-go-to distribution:')
print(paths_df["would_go_to"].value_counts().to_string())
```

### Terminus Diagnosis

**Primary attractor pair:**
- **Terminus:** `MEC LV VI Pyramidal Polymorphic` — 13 of 28 paths terminate here
- **Would-go-to:** `EC LIV VI Deep Multipolar Principal` — 13 of 28 paths would return here if revisits were allowed

This double-finding identifies `MEC LV VI Pyramidal Polymorphic` as the dominant structural
attractor of the excitatory scaffold. The signal is deterministically routed there and
then blocked — the only next step it wants to take is back to `EC LIV VI Deep Multipolar
Principal`, which was already visited. The two nodes form a trap-and-driver pair.

`MEC LV VI Pyramidal Polymorphic` also has a non-zero self-connection in the raw matrix,
consistent with its role as a recurrent self-reinforcing terminus.

## 9. Short Cycle Enumeration

Reciprocal pairs (length 2), triangles (length 3), and squares (length 4).
Weight product = product of all edge weights in the cycle. High weight product
indicates a cycle that carries the most recurrent signal.

The top-weight length-4 cycle validates the **trisynaptic circuit**.

```python
# Reciprocal pairs
reciprocal = [
    (u, v, G[u][v]['weight'] * G[v][u]['weight'])
    for u in G for v in G
    if u < v and G.has_edge(u,v) and G.has_edge(v,u)
]
reciprocal.sort(key=lambda x: -x[2])

# Triangles
triangles = []
for u in G:
    for v in G.successors(u):
        for w in G.successors(v):
            if w != u and G.has_edge(w, u):
                wt = G[u][v]['weight'] * G[v][w]['weight'] * G[w][u]['weight']
                triangles.append((u, v, w, wt))
triangles.sort(key=lambda x: -x[3])

# Squares
squares = []
for u in G:
    for v in G.successors(u):
        for w in G.successors(v):
            if w == u: continue
            for x in G.successors(w):
                if x not in (u, v, w) and G.has_edge(x, u):
                    wt = (G[u][v]['weight'] * G[v][w]['weight'] *
                          G[w][x]['weight'] * G[x][u]['weight'])
                    squares.append((u, v, w, x, wt))
squares.sort(key=lambda x: -x[4])

print(f'Reciprocal pairs : {len(reciprocal)}')
print(f'Triangles        : {len(triangles)}')
print(f'Squares          : {len(squares)}')

print(f'\nTop 5 reciprocal pairs (by weight product):')
for u,v,w in reciprocal[:5]:
    print(f'  {u} ↔ {v}  ({w:.3e})')

print(f'\nTop 5 triangles:')
for u,v,w,wt in triangles[:5]:
    print(f'  {u} → {v} → {w} → {u}  ({wt:.3e})')

print(f'\nTop 5 squares:')
for a,b,c,d,wt in squares[:5]:
    print(f'  {a} → {b} → {c} → {d} → {a}  ({wt:.3e})')

# Save
pd.DataFrame(reciprocal[:20], columns=['node1','node2','weight_product']
).to_csv(f'{TABLE_DIR}stage0_top_reciprocal.csv', index=False)
pd.DataFrame([(a,b,c,w) for a,b,c,w in triangles[:20]],
             columns=['n1','n2','n3','weight_product']
).to_csv(f'{TABLE_DIR}stage0_top_triangles.csv', index=False)
pd.DataFrame([(a,b,c,d,w) for a,b,c,d,w in squares[:20]],
             columns=['n1','n2','n3','n4','weight_product']
).to_csv(f'{TABLE_DIR}stage0_top_squares.csv', index=False)
```

**PageRank vs greedy terminus disagreement:** DG Mossy tops PageRank/Markov; MEC LV VI Pyramidal
Polymorphic tops greedy paths. This is structurally meaningful: random walkers recirculate through
high-degree hubs (DG Mossy has many incoming edges), while deterministic signal is trapped
at the local weight maximum that exhausts all unvisited neighbors. The two centrality families
are measuring different properties of the same network.
