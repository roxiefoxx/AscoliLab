# Stage 3 — Weighted Excitability Score for Connected Rees A–M Trimers

This notebook scores hippocampal 3-node motifs using a continuous weighted excitability score (wES). It fixes the previous notebook by using one consistent variable scheme from start to finish.

Workflow:
1. Detect Colab/local environment and set paths.
2. Load the signed weighted matrix `w_ij_gaa.csv`.
3. Load an existing full trimer table if available, otherwise generate all \(\binom{72}{3}\) trimers.
4. Filter to connected Rees A–M candidate trimers: at least two directed edges and all three nodes touching at least one edge.
5. Score each retained trimer with wES, Rees-style ES, weight summaries, and spectral metrics.
6. Generate interactive Plotly histograms and scatterplots.
7. Export all major tables and figures.


## 0. Environment Setup and Imports

This section mounts Google Drive only when running in Colab, then defines robust paths for either Colab or a local/Jupyter environment. The same path variables are used throughout the notebook.


```python
# ============================================================
# 0. Environment Setup, Imports, and Paths
# ============================================================

from pathlib import Path
from itertools import combinations
import ast
import os
import sys
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    import plotly.express as px
except ImportError as e:
    raise ImportError("Plotly is required for interactive figures. Install with: pip install plotly") from e

# ------------------------------------------------------------
# Detect Colab and mount Google Drive if needed
# ------------------------------------------------------------
IN_COLAB = "google.colab" in sys.modules

if IN_COLAB:
    from google.colab import drive
    DRIVE_MOUNT = Path("/content/drive")
    drive.mount(str(DRIVE_MOUNT))
    PROJECT_DIR = Path("/content/drive/MyDrive/Ascoli_LabRotation")
else:
    # Local/Jupyter fallback. Prefer /mnt/data when available in ChatGPT; otherwise use cwd.
    if Path("/mnt/data").exists():
        PROJECT_DIR = Path("/mnt/data")
    else:
        PROJECT_DIR = Path.cwd()

# ------------------------------------------------------------
# Data-path candidates
# ------------------------------------------------------------
WEIGHT_CANDIDATES = [
    PROJECT_DIR / "w_ij_gaa.csv",
    PROJECT_DIR / "matrices" / "w_ij_gaa.csv",
    Path("/mnt/data/w_ij_gaa.csv"),
]

TRIAD_CANDIDATES = [
    PROJECT_DIR / "stage2a_triad_enumeration_all.csv",
    PROJECT_DIR / "outputs" / "stage2a_triad_enumeration_all.csv",
    PROJECT_DIR / "trimer_audit" / "outputs" / "triads_enumeration.csv",
    Path("/mnt/data/stage2a_triad_enumeration_all.csv"),
]

def first_existing(candidates):
    for p in candidates:
        if p.exists() and p.is_file():
            return p
    return None

WEIGHT_PATH = first_existing(WEIGHT_CANDIDATES)
TRIAD_PATH = first_existing(TRIAD_CANDIDATES)

if WEIGHT_PATH is None:
    raise FileNotFoundError(
        "Could not find w_ij_gaa.csv. Checked:\n" +
        "\n".join(str(p) for p in WEIGHT_CANDIDATES)
    )

OUTPUT_DIR = PROJECT_DIR / "outputs_stage3_wES_fixed"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("IN_COLAB    :", IN_COLAB)
print("PROJECT_DIR :", PROJECT_DIR)
print("WEIGHT_PATH :", WEIGHT_PATH)
print("TRIAD_PATH  :", TRIAD_PATH if TRIAD_PATH else "None found; will generate all C(n,3) trimers")
print("OUTPUT_DIR  :", OUTPUT_DIR)
```

## 1. Load the Signed Weighted Connectivity Matrix

This section loads the 72 × 72 signed weighted matrix and validates that row/column labels align. Rows are treated as source/presynaptic neuron types and columns as target/postsynaptic neuron types.


```python
# ============================================================
# 1. Load Weighted Matrix
# ============================================================

A_df = pd.read_csv(WEIGHT_PATH, index_col=0)
A_df = A_df.apply(pd.to_numeric, errors="coerce").fillna(0.0)

A_df.index = A_df.index.astype(str).str.strip()
A_df.columns = A_df.columns.astype(str).str.strip()

# Reorder columns to match index if labels match as sets.
if set(A_df.index) == set(A_df.columns):
    A_df = A_df.loc[A_df.index, A_df.index]

if A_df.shape[0] != A_df.shape[1]:
    raise ValueError(f"Matrix is not square: {A_df.shape}")

if list(A_df.index) != list(A_df.columns):
    raise ValueError(
        "Row and column labels do not match. Check the weight matrix labels before proceeding."
    )

labels = list(A_df.index)
A = A_df.to_numpy(dtype=float)
N = len(labels)

print("Matrix shape:", A_df.shape)
print("Node count:", N)
print("All unordered trimers C(N,3):", len(list(combinations(labels, 3))))
print("Observed directed off-diagonal edges:", int(np.count_nonzero(A) - np.count_nonzero(np.diag(A))))

display(A_df.iloc[:5, :5])
```

## 2. Load or Generate the Full Trimer Table

This section loads the prior trimer enumeration if available. If no trimer table is found, it generates the complete unordered trimer list from the 72 matrix labels.

The notebook standardizes all trimer node columns to:
- `node_1`
- `node_2`
- `node_3`

If the trimer table contains explicit E/I identity columns (`ei_1`, `ei_2`, `ei_3`), those are preserved and used later for node signs.


```python
# ============================================================
# 2. Load or Generate Full Trimer List
# ============================================================

if TRIAD_PATH is not None and TRIAD_PATH.exists():
    df_triads_raw = pd.read_csv(TRIAD_PATH)
    print("Loaded trimer table:", TRIAD_PATH)
    print("Raw shape:", df_triads_raw.shape)
else:
    df_triads_raw = pd.DataFrame(combinations(labels, 3), columns=["node_1", "node_2", "node_3"])
    print("Generated full trimer table from matrix labels.")
    print("Raw shape:", df_triads_raw.shape)

# ------------------------------------------------------------
# Standardize node columns
# ------------------------------------------------------------
df_triads = df_triads_raw.copy()

rename_options = [
    ({"node1": "node_1", "node2": "node_2", "node3": "node_3"}, ["node1", "node2", "node3"]),
    ({"i": "node_1", "j": "node_2", "k": "node_3"}, ["i", "j", "k"]),
    ({"A": "node_1", "B": "node_2", "C": "node_3"}, ["A", "B", "C"]),
]

if all(c in df_triads.columns for c in ["node_1", "node_2", "node_3"]):
    pass
else:
    renamed = False
    for mapping, required in rename_options:
        if all(c in df_triads.columns for c in required):
            df_triads = df_triads.rename(columns=mapping)
            renamed = True
            break

    if not renamed and "nodes" in df_triads.columns:
        def parse_nodes(x):
            if isinstance(x, list):
                return x
            if isinstance(x, str):
                x = x.strip()
                if x.startswith("["):
                    return ast.literal_eval(x)
                if "|" in x:
                    return x.split("|")
                if "," in x:
                    return [part.strip() for part in x.split(",")]
            raise ValueError(f"Cannot parse nodes value: {x}")
        parsed = df_triads["nodes"].apply(parse_nodes)
        df_triads["node_1"] = parsed.apply(lambda x: x[0])
        df_triads["node_2"] = parsed.apply(lambda x: x[1])
        df_triads["node_3"] = parsed.apply(lambda x: x[2])
    elif not renamed:
        raise ValueError(
            "Could not find trimer node columns. Expected node_1/node_2/node_3, "
            "node1/node2/node3, i/j/k, A/B/C, or nodes."
        )

for c in ["node_1", "node_2", "node_3"]:
    df_triads[c] = df_triads[c].astype(str).str.strip()

# Keep all original metadata, but ensure the standardized columns are first.
front = ["node_1", "node_2", "node_3"]
rest = [c for c in df_triads.columns if c not in front]
df_triads = df_triads[front + rest]

# ------------------------------------------------------------
# Basic validation
# ------------------------------------------------------------
missing_from_matrix = sorted(
    set(df_triads["node_1"]).union(df_triads["node_2"]).union(df_triads["node_3"]) - set(labels)
)

if missing_from_matrix:
    print(f"Warning: {len(missing_from_matrix)} trimer labels are not present in A_df.")
    print(missing_from_matrix[:20])
else:
    print("All trimer labels are present in the weighted matrix.")

print("Standardized trimer table shape:", df_triads.shape)
display(df_triads.head())

# Export standardized input table
standardized_path = OUTPUT_DIR / "section2_standardized_input_trimers.csv"
df_triads.to_csv(standardized_path, index=False)
print("Saved:", standardized_path)
```

## 3. Filter to Connected Rees A–M Candidate Trimers

Rees superpatterns A–M are connected 3-node configurations. This section excludes disconnected superpatterns (`-A`, `-B`, `-C`) by requiring:

1. at least two observed directed edges inside the trimer, and  
2. all three nodes touch at least one edge.

This keeps simple chains and feedforward structures but removes isolated-node cases such as a reciprocal dyad plus a disconnected third node.


```python
# ============================================================
# 3. Filter Trimers: Connected Rees A–M Candidate Rule
# ============================================================

OBS_DIR = (A != 0)
np.fill_diagonal(OBS_DIR, False)
node_to_idx = {node: idx for idx, node in enumerate(labels)}

filter_records = []
keep_mask = []

for row_idx, row in df_triads.iterrows():
    triad_nodes = [row["node_1"], row["node_2"], row["node_3"]]

    if any(n not in node_to_idx for n in triad_nodes):
        filter_records.append({
            "row_index": row_idx,
            "n_directed_edges_in_trimer": np.nan,
            "n_nodes_touching_edge": np.nan,
            "rees_connected_A_to_M_candidate": False,
            "filter_status": "missing_node",
        })
        keep_mask.append(False)
        continue

    idx = [node_to_idx[n] for n in triad_nodes]
    sub_obs = OBS_DIR[np.ix_(idx, idx)]

    n_directed_edges = int(sub_obs.sum())
    node_touches_edge = sub_obs.any(axis=0) | sub_obs.any(axis=1)
    n_nodes_touching = int(node_touches_edge.sum())

    is_rees_connected_A_to_M = (
        n_directed_edges >= 2
        and n_nodes_touching == 3
    )

    filter_records.append({
        "row_index": row_idx,
        "n_directed_edges_in_trimer": n_directed_edges,
        "n_nodes_touching_edge": n_nodes_touching,
        "rees_connected_A_to_M_candidate": is_rees_connected_A_to_M,
        "filter_status": "ok",
    })
    keep_mask.append(is_rees_connected_A_to_M)

filter_df = pd.DataFrame(filter_records)

df_triads_filter_audit = pd.concat(
    [df_triads.reset_index(drop=True), filter_df.drop(columns=["row_index"]).reset_index(drop=True)],
    axis=1
)

df_triads_rees = (
    df_triads_filter_audit[df_triads_filter_audit["rees_connected_A_to_M_candidate"]]
    .reset_index(drop=True)
)

print("Original trimers:", len(df_triads))
print("Connected Rees A–M candidate trimers:", len(df_triads_rees))
print("Removed trimers:", len(df_triads) - len(df_triads_rees))

edge_touch_summary = (
    df_triads_filter_audit
    .groupby(["n_directed_edges_in_trimer", "n_nodes_touching_edge"], dropna=False)
    .size()
    .reset_index(name="trimer_count")
    .sort_values(["n_directed_edges_in_trimer", "n_nodes_touching_edge"])
)

display(edge_touch_summary)
display(df_triads_rees.head())

# Exports
filter_audit_path = OUTPUT_DIR / "section3_trimer_filter_audit_all.csv"
filtered_path = OUTPUT_DIR / "section3_connected_rees_A_to_M_trimers.csv"
edge_touch_path = OUTPUT_DIR / "section3_edge_touch_filter_summary.csv"

df_triads_filter_audit.to_csv(filter_audit_path, index=False)
df_triads_rees.to_csv(filtered_path, index=False)
edge_touch_summary.to_csv(edge_touch_path, index=False)

print("Saved:", filter_audit_path)
print("Saved:", filtered_path)
print("Saved:", edge_touch_path)
```

## 4. Define wES, Rees ES, and Spectral Helper Functions

The weighted excitability score is computed node-wise and summed across the trimer:

$$
wES_i = s_i \left(1 + \alpha \log(1 + E_i) - \beta \log(1 + I_i)\right)
$$

$$
wES_{\text{triad}} = \sum_{i=1}^{3} wES_i
$$

where \(s_i\) is the node E/I sign, \(E_i\) is incoming excitatory drive, and \(I_i\) is incoming inhibitory drive.

The spectral radius is:

$$
\rho(W_{\text{triad}}) = \max_i |\lambda_i|
$$

The Frobenius norm is:

$$
\|W_{\text{triad}}\|_F = \sqrt{\sum_i \sum_j w_{ij}^2}
$$


```python
# ============================================================
# 4. Helper Functions
# ============================================================

EI_MAP = {
    "e": 1, "exc": 1, "excitatory": 1, "+": 1, "+1": 1, 1: 1,
    "i": -1, "inh": -1, "inhibitory": -1, "-": -1, "-1": -1, -1: -1,
}

def normalize_ei_value(x):
    if pd.isna(x):
        return None
    key = str(x).strip().lower()
    return EI_MAP.get(key, None)


def infer_node_sign_from_outgoing(node, A_df):
    """
    Fallback E/I estimate used only if explicit E/I metadata is unavailable.
    +1 if outgoing signed weight sum is nonnegative; -1 if negative.
    """
    return 1 if A_df.loc[node].sum() >= 0 else -1


def get_node_signs(row, A_df):
    """
    Prefer explicit ei_1/ei_2/ei_3 columns when present.
    Otherwise infer from outgoing signed-weight balance.
    """
    triad_nodes = [row["node_1"], row["node_2"], row["node_3"]]

    if all(c in row.index for c in ["ei_1", "ei_2", "ei_3"]):
        signs = [normalize_ei_value(row[c]) for c in ["ei_1", "ei_2", "ei_3"]]
        if all(s in [-1, 1] for s in signs):
            return signs, "explicit_ei_columns"

    signs = [infer_node_sign_from_outgoing(n, A_df) for n in triad_nodes]
    return signs, "outgoing_weight_inference"


def weighted_es(subW, node_signs, alpha=0.10, beta=0.10):
    """
    Weighted excitability score.

    For each node:
    - baseline sign is E/I identity
    - incoming excitation increases score
    - incoming inhibition decreases score
    - log1p handles extreme magnitude ranges
    """
    node_scores = []

    for target_idx in range(3):
        incoming = subW[:, target_idx].copy()
        incoming[target_idx] = 0.0

        exc_drive = incoming[incoming > 0].sum()
        inh_drive = np.abs(incoming[incoming < 0]).sum()

        s_i = node_signs[target_idx]
        score = s_i * (
            1
            + alpha * np.log1p(exc_drive)
            - beta * np.log1p(inh_drive)
        )
        node_scores.append(float(score))

    return float(np.sum(node_scores)), node_scores


def rees_style_es(subW, node_signs):
    """
    Rees-style categorical ES approximation.

    Baseline:
    +1 excitatory node
    -1 inhibitory node

    Incoming excitatory edge: multiply by 1.1
    Incoming inhibitory edge: multiply by 0.9
    """
    node_scores = []

    for target_idx in range(3):
        score = float(node_signs[target_idx])

        for source_idx in range(3):
            if source_idx == target_idx:
                continue

            w = subW[source_idx, target_idx]
            if w > 0:
                score *= 1.1
            elif w < 0:
                score *= 0.9

        node_scores.append(float(score))

    return float(np.sum(node_scores)), node_scores


def spectral_metrics(subW):
    eigvals = np.linalg.eigvals(subW)
    abs_eigvals = np.abs(eigvals)
    dominant_idx = int(np.argmax(abs_eigvals))
    dominant = eigvals[dominant_idx]

    complex_eig_count = int(np.sum(np.abs(np.imag(eigvals)) > 1e-9))

    return {
        "spectral_radius": float(abs_eigvals[dominant_idx]),
        "frobenius_norm": float(np.linalg.norm(subW, ord="fro")),
        "determinant": float(np.linalg.det(subW)),
        "dominant_eigval_real": float(np.real(dominant)),
        "dominant_eigval_imag": float(np.imag(dominant)),
        "eigval_1_real": float(np.real(eigvals[0])),
        "eigval_1_imag": float(np.imag(eigvals[0])),
        "eigval_2_real": float(np.real(eigvals[1])),
        "eigval_2_imag": float(np.imag(eigvals[1])),
        "eigval_3_real": float(np.real(eigvals[2])),
        "eigval_3_imag": float(np.imag(eigvals[2])),
        "complex_eig_count": complex_eig_count,
        "has_complex_eigs": bool(complex_eig_count > 0),
    }
```

## 5. Score Connected Trimers

This section scores each connected Rees A–M candidate trimer with:
- wES
- Rees-style ES
- total positive, negative, and absolute weight
- directed edge count
- spectral radius
- Frobenius norm
- determinant and eigenvalue summaries


```python
# ============================================================
# 5. Score Connected Trimers
# ============================================================

# This cell is written with NumPy arrays instead of DataFrame.iterrows()
# because 20k+ trimers with eigendecomposition can otherwise be slow in Colab.

records = []

node1_arr = df_triads_rees["node_1"].astype(str).to_numpy()
node2_arr = df_triads_rees["node_2"].astype(str).to_numpy()
node3_arr = df_triads_rees["node_3"].astype(str).to_numpy()

# Optional metadata preserved from previous enumeration, if available.
optional_cols = [
    c for c in ["superpattern", "superpattern_name", "ei_pattern", "is_connected", "n_edges", "total_abs_weight"]
    if c in df_triads_rees.columns
]
optional_data = {c: df_triads_rees[c].to_numpy() for c in optional_cols}

# Prefer explicit E/I columns if available.
has_explicit_ei = all(c in df_triads_rees.columns for c in ["ei_1", "ei_2", "ei_3"])
if has_explicit_ei:
    ei1_arr = df_triads_rees["ei_1"].to_numpy()
    ei2_arr = df_triads_rees["ei_2"].to_numpy()
    ei3_arr = df_triads_rees["ei_3"].to_numpy()

# Cache fallback signs so each node is inferred once.
fallback_sign_cache = {node: infer_node_sign_from_outgoing(node, A_df) for node in labels}

for triad_index in range(len(df_triads_rees)):
    triad_nodes = [node1_arr[triad_index], node2_arr[triad_index], node3_arr[triad_index]]

    missing = [n for n in triad_nodes if n not in node_to_idx]
    if missing:
        records.append({
            "triad_index": triad_index,
            "node_1": triad_nodes[0],
            "node_2": triad_nodes[1],
            "node_3": triad_nodes[2],
            "status": "failed_missing_node",
            "missing_nodes": "|".join(missing),
        })
        continue

    idx = [node_to_idx[n] for n in triad_nodes]
    subW = A[np.ix_(idx, idx)].astype(float, copy=True)
    np.fill_diagonal(subW, 0.0)

    if has_explicit_ei:
        node_signs = [
            normalize_ei_value(ei1_arr[triad_index]),
            normalize_ei_value(ei2_arr[triad_index]),
            normalize_ei_value(ei3_arr[triad_index]),
        ]
        if not all(s in [-1, 1] for s in node_signs):
            node_signs = [fallback_sign_cache[n] for n in triad_nodes]
            sign_source = "outgoing_weight_inference_after_bad_ei"
        else:
            sign_source = "explicit_ei_columns"
    else:
        node_signs = [fallback_sign_cache[n] for n in triad_nodes]
        sign_source = "outgoing_weight_inference"

    wes_total, wes_node_scores = weighted_es(subW, node_signs)
    rees_total, rees_node_scores = rees_style_es(subW, node_signs)
    spec = spectral_metrics(subW)

    pos_weight_sum = float(subW[subW > 0].sum())
    neg_weight_sum = float(subW[subW < 0].sum())
    abs_weight_sum = float(np.abs(subW).sum())
    nonzero_edges = int(np.count_nonzero(subW))

    rec = {
        "triad_index": triad_index,
        "node_1": triad_nodes[0],
        "node_2": triad_nodes[1],
        "node_3": triad_nodes[2],
        "status": "ok",
        "sign_source": sign_source,

        "node_1_sign": node_signs[0],
        "node_2_sign": node_signs[1],
        "node_3_sign": node_signs[2],
        "ei_signature": "".join(["E" if s == 1 else "I" for s in node_signs]),

        "wES": wes_total,
        "wES_node_1": wes_node_scores[0],
        "wES_node_2": wes_node_scores[1],
        "wES_node_3": wes_node_scores[2],

        "rees_ES": rees_total,
        "rees_node_1": rees_node_scores[0],
        "rees_node_2": rees_node_scores[1],
        "rees_node_3": rees_node_scores[2],

        "wES_minus_rees_ES": wes_total - rees_total,
        "abs_wES_minus_rees_ES": abs(wes_total - rees_total),

        "positive_weight_sum": pos_weight_sum,
        "negative_weight_sum": neg_weight_sum,
        "absolute_weight_sum": abs_weight_sum,
        "nonzero_directed_edges": nonzero_edges,
    }

    for c in optional_cols:
        rec[f"input_{c}"] = optional_data[c][triad_index]

    rec.update(spec)
    records.append(rec)

    if (triad_index + 1) % 5000 == 0:
        print(f"Scored {triad_index + 1:,} / {len(df_triads_rees):,} trimers")

df_wES_full = pd.DataFrame(records)
df_ok = df_wES_full[df_wES_full["status"] == "ok"].copy()

print("Scored table shape:", df_wES_full.shape)
print("Valid scored trimers:", len(df_ok))
print("Failed trimers:", len(df_wES_full) - len(df_ok))

display(df_wES_full.head())

# Exports
scores_path = OUTPUT_DIR / "section5_connected_trimer_wES_scores.csv"
valid_scores_path = OUTPUT_DIR / "section5_valid_connected_trimer_wES_scores.csv"

df_wES_full.to_csv(scores_path, index=False)
df_ok.to_csv(valid_scores_path, index=False)

print("Saved:", scores_path)
print("Saved:", valid_scores_path)
```

## 6. Rank Trimers

This section creates ranked tables for the strongest trimers by wES, lowest wES, largest absolute synaptic drive, largest spectral radius, and largest divergence between categorical Rees ES and weighted ES.


```python
# ============================================================
# 6. Rank Trimers
# ============================================================

TOP_N = 100

df_top_wES = df_ok.sort_values("wES", ascending=False).head(TOP_N)
df_bottom_wES = df_ok.sort_values("wES", ascending=True).head(TOP_N)
df_top_abs_drive = df_ok.sort_values("absolute_weight_sum", ascending=False).head(TOP_N)
df_top_spectral = df_ok.sort_values("spectral_radius", ascending=False).head(TOP_N)
df_largest_shift = df_ok.sort_values("abs_wES_minus_rees_ES", ascending=False).head(TOP_N)

print("Top 20 by wES")
display(df_top_wES.head(20))

print("Top 20 by spectral radius")
display(df_top_spectral.head(20))

# Exports
df_top_wES.to_csv(OUTPUT_DIR / "section6_top100_highest_wES.csv", index=False)
df_bottom_wES.to_csv(OUTPUT_DIR / "section6_top100_lowest_wES.csv", index=False)
df_top_abs_drive.to_csv(OUTPUT_DIR / "section6_top100_absolute_drive.csv", index=False)
df_top_spectral.to_csv(OUTPUT_DIR / "section6_top100_spectral_radius.csv", index=False)
df_largest_shift.to_csv(OUTPUT_DIR / "section6_top100_largest_wES_rees_shift.csv", index=False)

print("Section 6 ranked-table exports complete.")
```

## 9. Summary by E/I Signature and Optional Superpattern

This section summarizes score distributions by E/I node composition. If the input trimer table included Rees superpattern labels, it also summarizes by superpattern.


```python
# ============================================================
# 9. Group Summaries
# ============================================================

summary_by_ei = (
    df_ok
    .groupby("ei_signature")
    .agg(
        n_trimers=("triad_index", "count"),
        mean_wES=("wES", "mean"),
        median_wES=("wES", "median"),
        mean_rees_ES=("rees_ES", "mean"),
        median_rees_ES=("rees_ES", "median"),
        mean_abs_drive=("absolute_weight_sum", "mean"),
        median_abs_drive=("absolute_weight_sum", "median"),
        mean_spectral_radius=("spectral_radius", "mean"),
        median_spectral_radius=("spectral_radius", "median"),
        complex_fraction=("has_complex_eigs", "mean"),
    )
    .reset_index()
    .sort_values("n_trimers", ascending=False)
)

display(summary_by_ei)
summary_by_ei.to_csv(OUTPUT_DIR / "section9_summary_by_ei_signature.csv", index=False)

# Optional superpattern summary if present from input.
superpattern_col = None
for candidate in ["input_superpattern", "superpattern"]:
    if candidate in df_ok.columns:
        superpattern_col = candidate
        break

if superpattern_col is not None:
    summary_by_superpattern = (
        df_ok
        .groupby(superpattern_col)
        .agg(
            n_trimers=("triad_index", "count"),
            mean_wES=("wES", "mean"),
            median_wES=("wES", "median"),
            mean_abs_drive=("absolute_weight_sum", "mean"),
            mean_spectral_radius=("spectral_radius", "mean"),
        )
        .reset_index()
        .sort_values("n_trimers", ascending=False)
    )
    display(summary_by_superpattern)
    summary_by_superpattern.to_csv(OUTPUT_DIR / "section9_summary_by_superpattern.csv", index=False)
else:
    print("No superpattern column found; skipped superpattern summary.")

print("Section 9 summaries saved.")
```

## 10. Node Participation in High-Scoring Trimers

This section identifies neuron types that repeatedly appear in the highest-wES trimers. This helps separate motif-level effects from individual hub/neuron-type effects.


```python
# ============================================================
# 10. Node Participation in High-Scoring Trimers
# ============================================================

TOP_N = 100
top_df = df_ok.sort_values("wES", ascending=False).head(TOP_N)

participation_records = []
for _, row in top_df.iterrows():
    for node_col in ["node_1", "node_2", "node_3"]:
        participation_records.append({
            "node": row[node_col],
            "triad_index": row["triad_index"],
            "wES": row["wES"],
            "spectral_radius": row["spectral_radius"],
            "absolute_weight_sum": row["absolute_weight_sum"],
            "ei_signature": row["ei_signature"],
        })

df_top_participation = pd.DataFrame(participation_records)

node_participation_summary = (
    df_top_participation
    .groupby("node")
    .agg(
        top_trimer_count=("triad_index", "count"),
        mean_top_wES=("wES", "mean"),
        max_top_wES=("wES", "max"),
        mean_top_spectral_radius=("spectral_radius", "mean"),
        mean_top_abs_drive=("absolute_weight_sum", "mean"),
    )
    .reset_index()
    .sort_values("top_trimer_count", ascending=False)
)

display(node_participation_summary.head(30))

participation_path = OUTPUT_DIR / "section10_node_participation_top_wES_trimers.csv"
node_participation_summary.to_csv(participation_path, index=False)
print("Saved:", participation_path)
```

## 11. Final Master Exports

This final section exports the master scored table, the valid scored subset, ranked tables, summaries, and a compact run summary.


```python
# ============================================================
# 11. Final Master Exports
# ============================================================

master_path = OUTPUT_DIR / "FINAL_full_connected_rees_A_to_M_wES_spectral_table.csv"
valid_path = OUTPUT_DIR / "FINAL_valid_connected_rees_A_to_M_wES_spectral_table.csv"
summary_path = OUTPUT_DIR / "FINAL_summary_metrics.csv"

# Final master tables
df_wES_full.to_csv(master_path, index=False)
df_ok.to_csv(valid_path, index=False)

summary = pd.DataFrame({
    "metric": [
        "input_trimers_before_filter",
        "connected_rees_A_to_M_candidate_trimers",
        "valid_scored_trimers",
        "failed_scored_trimers",
        "mean_wES",
        "median_wES",
        "mean_rees_ES",
        "median_rees_ES",
        "mean_spectral_radius",
        "median_spectral_radius",
        "mean_absolute_weight_sum",
        "median_absolute_weight_sum",
    ],
    "value": [
        len(df_triads),
        len(df_triads_rees),
        len(df_ok),
        len(df_wES_full) - len(df_ok),
        df_ok["wES"].mean(),
        df_ok["wES"].median(),
        df_ok["rees_ES"].mean(),
        df_ok["rees_ES"].median(),
        df_ok["spectral_radius"].mean(),
        df_ok["spectral_radius"].median(),
        df_ok["absolute_weight_sum"].mean(),
        df_ok["absolute_weight_sum"].median(),
    ]
})

summary.to_csv(summary_path, index=False)
display(summary)

print("Final exports complete.")
print("Master table:", master_path)
print("Valid table :", valid_path)
print("Summary     :", summary_path)
print("Output dir  :", OUTPUT_DIR)
```
